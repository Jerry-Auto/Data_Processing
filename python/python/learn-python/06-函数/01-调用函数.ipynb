# 关键字参数 **kw

使用 `**kw` 可以将关键字参数组装成 dict。

```python
# 关键字参数
def person(name, age, **kw):
    """
    可以接受任意数量的关键字参数
    关键字参数会被组装成 dict
    """
    print('name:', name, 'age:', age, 'others:', kw)

# 可以只传入必需参数
person('Michael', 30)

# 也可以传入任意数量的关键字参数
person('Bob', 25, city='Beijing')
person('Adam', 30, city='Shanghai', job='Engineer')
```

```python
# 传入 dict 作为关键字参数
extra = {'city': 'Beijing', 'job': 'Engineer'}
person('Jack', 24, **extra)

# **extra 表示把 extra dict 的所有 key-value 作为关键字参数传入
```
