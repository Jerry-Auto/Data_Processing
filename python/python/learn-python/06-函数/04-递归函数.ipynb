# 递归函数 - 汉诺塔问题

经典的递归问题：将 n 个盘子从 A 柱移动到 C 柱，可以借助 B 柱，每次只能移动一个盘子，大盘子不能放在小盘子上面。

```python
def move(n, a, b, c):
    """
    汉诺塔移动函数
    
    参数:
        n: 盘子数量
        a: 起始柱
        b: 辅助柱
        c: 目标柱
    """
    if n == 1:
        # 只有一个盘子时，直接移动
        print(a, "-->", c)
    else:
        # 1. 将 n-1 个盘子从 a 移到 b (借助 c)
        move(n-1, a, c, b)
        # 2. 将第 n 个盘子从 a 移到 c
        print(a, "-->", c)
        # 3. 将 n-1 个盘子从 b 移到 c (借助 a)
        move(n-1, b, a, c)

# 测试：移动 3 个盘子
move(3, 'A', 'B', 'C')
```
