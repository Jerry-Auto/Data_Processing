# 函数作用域 (LEGB 规则)

Python 中变量作用域的查找顺序：Local → Enclosing → Global → Built-in

## Local - 局部作用域

```python
# 局部变量只在函数内部有效
def local_example():
    local_var = '我是局部变量'
    print(local_var)

local_example()
# print(local_var)  # NameError: 函数外部无法访问
```

## Enclosing - 嵌套作用域

```python
# 内层函数可以访问外层函数的变量
def outer():
    outer_var = '外层变量'
    
    def inner():
        print(f'内层访问：{outer_var}')
    
    inner()

outer()
```

## Global - 全局作用域

```python
# 全局变量在整个模块中有效
global_var = '我是全局变量'

def show_global():
    print(f'函数内访问：{global_var}')

show_global()
print(f'函数外访问：{global_var}')
```

```python
# 使用 global 关键字修改全局变量
count = 0

def increment():
    global count
    count += 1
    print(f'count = {count}')

increment()
increment()
print(f'最终 count = {count}')
```

## nonlocal - 修改嵌套作用域变量

```python
# nonlocal 用于修改外层函数的变量
def outer():
    count = 0
    
    def inner():
        nonlocal count
        count += 1
        print(f'inner: count = {count}')
    
    inner()
    inner()
    print(f'outer: count = {count}')

outer()
```

## Built-in - 内建作用域

```python
# 内建函数在最外层
print(len([1, 2, 3]))  # len 是内建函数
print(sum([1, 2, 3]))  # sum 是内建函数

# 可以查看内建作用域的内容
import builtins
print('内建函数数量:', len(dir(builtins)))
```

## 作用域查找顺序示例

```python
# LEGB 顺序示例
x = 'global'  # Global

def outer():
    x = 'enclosing'  # Enclosing
    
    def inner():
        x = 'local'  # Local
        print(f'inner: {x}')
    
    inner()
    print(f'outer: {x}')

outer()
print(f'global: {x}')
```
