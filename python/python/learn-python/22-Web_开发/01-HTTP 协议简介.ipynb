# HTTP 协议简介

HTTP 是超文本传输协议，是 Web 开发的基础。

## HTTP 请求结构

```python
# HTTP 请求由以下部分组成：
# 1. 请求行：方法 URL HTTP 版本
# 2. 请求头：键值对
# 3. 空行
# 4. 请求体（可选）

# GET 请求示例
get_request = '''GET /index.html HTTP/1.1
Host: www.example.com
User-Agent: Mozilla/5.0
Accept: text/html

'''

# POST 请求示例
post_request = '''POST /login HTTP/1.1
Host: www.example.com
Content-Type: application/x-www-form-urlencoded
Content-Length: 23

username=admin&password=123
'''

print('HTTP 请求结构示例')
print('GET 请求包含：请求行 + 请求头 + 空行')
print('POST 请求 additionally 包含：请求体')
```

## HTTP 响应结构

```python
# HTTP 响应由以下部分组成：
# 1. 状态行：HTTP 版本 状态码 状态描述
# 2. 响应头：键值对
# 3. 空行
# 4. 响应体

http_response = '''HTTP/1.1 200 OK
Date: Mon, 27 Jul 2026 12:00:00 GMT
Content-Type: text/html
Content-Length: 138

<html>
<head><title>Example</title></head>
<body><h1>Hello, World!</h1></body>
</html>
'''

print('HTTP 响应结构示例')
print('状态行：HTTP/1.1 200 OK')
print('常见状态码：200(OK), 301(重定向), 404(未找到), 500(服务器错误)')
```

## 常见 HTTP 方法

```python
http_methods = {
    'GET': '获取资源',
    'POST': '提交数据',
    'PUT': '更新资源',
    'DELETE': '删除资源',
    'HEAD': '获取响应头',
    'OPTIONS': '获取支持的方法',
}

for method, desc in http_methods.items():
    print(f'{method}: {desc}')
```
