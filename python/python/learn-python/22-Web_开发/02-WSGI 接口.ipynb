# WSGI 接口

WSGI 是 Python Web 应用的标准接口。

```python
# WSGI 应用是一个可调用对象
# 接收 environ 和 start_response 两个参数

def application(environ, start_response):
    """最简单的 WSGI 应用"""
    start_response('200 OK', [('Content-Type', 'text/html')])
    return [b'<h1>Hello, Web!</h1>']

print('WSGI 应用结构:')
print('def application(environ, start_response):')
print('    start_response(status, headers)')
print('    return [body]')
```

```python
# 根据 URL 返回不同内容
def application(environ, start_response):
    start_response('200 OK', [('Content-Type', 'text/html')])
    
    path = environ.get('PATH_INFO', '/')
    
    if path == '/':
        body = b'<h1>首页</h1>'
    elif path == '/signin':
        body = b'<h1>登录页面</h1>'
    else:
        body = b'<h1>404 Not Found</h1>'
    
    return [body]

print('根据 PATH_INFO 路由到不同处理函数')
```
