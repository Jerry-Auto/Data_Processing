# 操作文件和目录

使用 os 和 pathlib 模块操作文件和目录。

```python
import os

# 获取当前工作目录
print('当前目录:', os.getcwd())

# 获取绝对路径
print('绝对路径:', os.path.abspath('.'))
```

```python
# 路径拼接（推荐方式）
path = os.path.join(os.getcwd(), 'test_dir')
print('拼接路径:', path)

# 创建目录
if not os.path.exists(path):
    os.mkdir(path)
    print('已创建目录:', path)
else:
    print('目录已存在:', path)
```

```python
# 删除目录
path = os.path.join(os.getcwd(), 'test_dir')
if os.path.exists(path):
    os.rmdir(path)
    print('已删除目录:', path)
```

```python
# 拆分路径
path = '/home/user/test.txt'
print('splitext("/home/user/test.txt"):', os.path.splitext(path))
# ('/home/user/test', '.txt')

# 获取文件名
print('basename:', os.path.basename(path))

# 获取目录名
print('dirname:', os.path.dirname(path))
```

```python
# 列出目录内容
print('当前目录内容:', os.listdir('.'))
```
