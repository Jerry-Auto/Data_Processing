# StringIO 和 BytesIO

在内存中读写字符串和二进制数据。

## StringIO

```python
from io import StringIO

# 创建一个 StringIO 对象
f = StringIO()

# 写入字符串
f.write('Hello')
f.write(' ')
f.write('World')

# 获取写入的内容
print('f.getvalue():', f.getvalue())

# 从 StringIO 读取
f = StringIO('Hello\nWorld\n!')
print('read():')
while True:
    s = f.readline()
    if s == '':
        break
    print(s.strip())
```

## BytesIO

```python
from io import BytesIO

# 创建 BytesIO 对象
f = BytesIO()

# 写入 bytes
f.write('中文'.encode('utf-8'))

# 获取内容
print('f.getvalue():', f.getvalue())

# 从 BytesIO 读取
f = BytesIO(b'\xe4\xb8\xad\xe6\x96\x87')
print('read():', f.read())
print('decode:', f.getvalue().decode('utf-8'))
```
