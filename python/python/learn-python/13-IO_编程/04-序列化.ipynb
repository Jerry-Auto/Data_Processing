# 序列化

将对象转换为可存储或传输的格式（如 JSON）。

## JSON 序列化

```python
import json

# Python 对象
d = dict(name='Bob', age=20, score=88)

# 序列化为 JSON 字符串
json_str = json.dumps(d)
print('json_str:', json_str)

# 反序列化为 Python 对象
d2 = json.loads(json_str)
print('d2:', d2)
print('type(d2):', type(d2))
```

## JSON 序列化类实例

```python
import json

class Student(object):
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

s = Student('Bob', 20, 88)

# 使用 default 函数将类实例转换为 dict
def student2dict(std):
    return {
        'name': std.name,
        'age': std.age,
        'score': std.score
    }

json_str = json.dumps(s, default=student2dict)
print('json_str:', json_str)

# 反序列化
def dict2student(d):
    return Student(d['name'], d['age'], d['score'])

s2 = json.loads(json_str, object_hook=dict2student)
print('s2:', s2)
print('s2.name:', s2.name)
```
