# 使用模块

Python 的模块是使用 `.py` 文件组织的，每个 `.py` 文件就是一个模块。

## 导入模块

```python
# 导入模块的几种方式

# 方式 1: 导入整个模块
import math
print('sqrt(4):', math.sqrt(4))

# 方式 2: 从模块导入特定函数
from math import sqrt, sin
print('sqrt(9):', sqrt(9))

# 方式 3: 导入并重命名
import numpy as np
```

## 定义自己的模块

```python
# 模块可以包含函数、类、变量等
# 使用 if __name__ == '__main__' 可以在被导入时不执行测试代码

# 这是一个模块示例（实际应保存为 separate .py file）

def greet(name):
    """向指定的人打招呼"""
    return f'Hello, {name}!'

def farewell(name):
    """向指定的人告别"""
    return f'Goodbye, {name}!'

# 测试代码 - 只有直接运行此文件时才会执行
if __name__ == '__main__':
    print(greet('World'))
    print(farewell('World'))
```

## 模块的搜索路径

```python
import sys

# 查看模块搜索路径
print('模块搜索路径:')
for path in sys.path:
    print(f'  {path}')
```

## 使用别名避免命名冲突

```python
# 如果两个模块有相同名字的函数，可以使用别名
from math import log as log_math

# 假设有另一个模块也有 log 函数
# from other_module import log as log_other

print('log_math(10):', log_math(10))
```
