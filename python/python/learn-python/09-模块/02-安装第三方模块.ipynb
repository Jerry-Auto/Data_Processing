# 安装第三方模块

Python 有大量的第三方模块可供使用，通过 pip 工具安装。

## 使用 pip 安装模块

```python
# 在命令行中使用 pip 安装第三方模块
# pip install <module_name>

# 例如：
# pip install requests    # 安装 requests 库
# pip install numpy       # 安装 numpy 库
# pip install pandas      # 安装 pandas 库

# 查看已安装的模块
# pip list

# 卸载模块
# pip uninstall <module_name>

print('pip 是 Python 的包管理工具')
print('用于安装、升级和删除第三方模块')
```

## 使用 requirements.txt

```python
# 在项目中，通常使用 requirements.txt 文件记录依赖
# 文件格式示例:
# requests==2.28.0
# numpy>=1.21.0
# pandas

# 安装所有依赖:
# pip install -r requirements.txt

# 导出已安装的模块列表:
# pip freeze > requirements.txt

print('requirements.txt 用于记录项目依赖的第三方模块')
```
