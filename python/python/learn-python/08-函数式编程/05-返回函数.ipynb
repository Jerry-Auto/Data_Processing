# 返回函数 - 闭包

演示闭包的用法：函数内部定义函数，并返回内部函数。

```python
def createCounter():
    """
    创建一个计数器函数
    每次调用返回的计数器，计数加 1
    """
    count = 0

    def counter():
        nonlocal count  # 使用外部函数的变量
        count += 1
        return count

    return counter

# 测试
counterA = createCounter()
print('counterA:', counterA(), counterA(), counterA(), counterA(), counterA())  # 1 2 3 4 5

counterB = createCounter()
print('counterB:', [counterB(), counterB(), counterB(), counterB()])  # [1, 2, 3, 4]

if [counterB(), counterB(), counterB(), counterB()] == [1, 2, 3, 4]:
    print('测试通过!')
else:
    print('测试失败!')
```

```python
def createCounter2():
    """
    创建一个计数器函数
    每次调用返回的计数器，计数加 1
    """
    funcs=[]

    for i in range(1, 4):
        def counter():
            return i
        funcs.append(counter)

    return funcs
```

```python
for f in createCounter2():
    print(f())  # 1 2 3
```

```python
def createCounter3():
    """
    创建一个计数器函数
    每次调用返回的计数器，计数加 1
    """
    funcs=[]

    for i in range(1, 4):
        
        def counter(num):  # 使用默认参数绑定当前的 i

            def inner():
                return num
            
            return inner
        funcs.append(counter(i))

    return funcs
```

```python
for f in createCounter3():
    print(f())  # 1 2 3
```
