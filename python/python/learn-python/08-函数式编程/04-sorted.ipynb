# sorted - 排序

演示 `sorted()` 函数的用法，包括自定义排序规则。

```python
# 按名字排序
L = [('Bob', 75), ('Adam', 92), ('Bart', 66), ('Lisa', 88)]

def by_name(t):
    """按名字排序"""
    return t[0]

def by_score(t):
    """按分数排序"""
    return t[1]

# 按名字升序排序
L2 = sorted(L, key=by_name)
print('按名字排序:', L2)

# 按分数降序排序
L3 = sorted(L, key=by_score, reverse=True)
print('按分数降序:', L3)
```
