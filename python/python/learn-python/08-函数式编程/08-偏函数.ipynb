# 偏函数

使用 `functools.partial` 可以固定函数的某些参数，创建新函数。

```python
# int() 函数可以带 base 参数进行进制转换
print('int("12345"):', int('12345'))  # 默认按十进制
print('int("12345", base=8):', int('12345', base=8))  # 按八进制
print('int("12345", base=16):', int('12345', base=16))  # 按十六进制
```

```python
import functools

# 使用偏函数固定 base=2，创建一个专门转换二进制的函数
int2 = functools.partial(int, base=2)

print('int2("1000000"):', int2('1000000'))  # 32
print('int2("1010101"):', int2('1010101'))  # 85
```

```python
# 也可以固定关键字参数
def add(a, b, c=0):
    return a + b + c

# 固定 c=10
add_with_10 = functools.partial(add, c=10)

print('add_with_10(1, 2):', add_with_10(1, 2))  # 13
print('add_with_10(5, 5):', add_with_10(5, 5))  # 20
```

```python
# 也可以固定位置参数
multiply = functools.partial(add, 2)  # 固定 a=2

print('multiply(3):', multiply(3))  # 5 (2+3+0)
print('multiply(3, c=5):', multiply(3, c=5))  # 10 (2+3+5)
```

## 小结

- `functools.partial` 可以固定原函数的某些参数
- 返回的新函数可以更简单地调用
- 适用于需要重复调用同一函数且部分参数固定的场景
