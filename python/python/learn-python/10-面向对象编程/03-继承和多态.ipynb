# 继承和多态

面向对象编程的重要特性：继承和多态。

## 继承

```python
# 定义父类
class Animal(object):
    def run(self):
        print('Animal is running...')

# 定义子类
class Dog(Animal):
    pass

class Cat(Animal):
    pass

# 子类继承父类的方法
dog = Dog()
dog.run()  # Animal is running...

cat = Cat()
cat.run()  # Animal is running...
```

```python
# 子类可以重写父类的方法
class Dog(Animal):
    def run(self):
        print('Dog is running...')

class Cat(Animal):
    def run(self):
        print('Cat is running...')

dog = Dog()
dog.run()  # Dog is running...

cat = Cat()
cat.run()  # Cat is running...
```

## 多态

```python
# 多态：父类引用指向子类实例
def run_twice(animal):
    """接收 Animal 类型实例，调用 run 方法两次"""
    animal.run()
    animal.run()

# 传入 Dog 实例
run_twice(Dog())
# Dog is running...
# Dog is running...

# 传入 Cat 实例
run_twice(Cat())
# Cat is running...
# Cat is running...
```

```python
# 多态的优势：可以传入任意 Animal 的子类
# 开闭原则：对扩展开放，对修改关闭

class Tortoise(Animal):
    def run(self):
        print('Tortoise is running slowly...')

# 无需修改 run_twice 函数，直接可以使用
run_twice(Tortoise())
# Tortoise is running slowly...
# Tortoise is running slowly...
```

## isinstance() 判断类型

```python
# isinstance() 可以判断继承关系
dog = Dog()

print('isinstance(dog, Dog):', isinstance(dog, Dog))  # True
print('isinstance(dog, Animal):', isinstance(dog, Animal))  # True
print('isinstance(Animal(), Dog):', isinstance(Animal(), Dog))  # False

# 优先使用 isinstance() 而不是 type()
```
