# 多态 (Polymorphism)

面向对象编程的三大特性之一：多态，同一接口不同实现。

## 多态的概念

```python
# 多态：同一方法在不同对象上有不同行为

class Dog:
    def speak(self):
        return '汪汪汪!'

class Cat:
    def speak(self):
        return '喵喵喵!'

class Duck:
    def speak(self):
        return '嘎嘎嘎!'

# 统一的接口
def animal_sound(animal):
    print(animal.speak())

dog = Dog()
cat = Cat()
duck = Duck()

animal_sound(dog)   # 汪汪汪!
animal_sound(cat)   # 喵喵喵!
animal_sound(duck)  # 嘎嘎嘎!
```

## 继承实现多态

```python
# 通过继承和方法重写实现多态

class Animal:
    def speak(self):
        return '动物叫声'

class Dog(Animal):
    def speak(self):
        return '汪汪汪!'

class Cat(Animal):
    def speak(self):
        return '喵喵喵!'

# 多态演示
animals = [Dog(), Cat(), Animal()]

for animal in animals:
    print(f'{type(animal).__name__}: {animal.speak()}')
```

## 鸭子类型 (Duck Typing)

```python
# Python 的鸭子类型：不关心类型，只关心方法
# "如果它走起来像鸭子，叫起来像鸭子，那它就是鸭子"

class Timer:
    def start(self):
        print('计时开始')
    def stop(self):
        print('计时结束')

class Engine:
    def start(self):
        print('引擎启动')
    def stop(self):
        print('引擎关闭')

# 统一的处理函数，不关心对象类型
def operate(machine):
    machine.start()
    machine.stop()

operate(Timer())
operate(Engine())
```

## 多态的实际应用

```python
# 多态在实际编程中的应用

class Payment:
    def pay(self, amount):
        raise NotImplementedError

class Alipay(Payment):
    def pay(self, amount):
        print(f'使用支付宝支付{amount}元')

class WechatPay(Payment):
    def pay(self, amount):
        print(f'使用微信支付{amount}元')

class CreditCard(Payment):
    def pay(self, amount):
        print(f'使用信用卡支付{amount}元')

# 处理支付，不关心具体支付方式
def process_payment(payment_method, amount):
    payment_method.pay(amount)

process_payment(Alipay(), 100)
process_payment(WechatPay(), 200)
process_payment(CreditCard(), 300)
```
