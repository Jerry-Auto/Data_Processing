# 获取对象信息

使用 type()、isinstance()、dir() 等函数获取对象信息。

## 使用 type()

```python
# 判断基本类型
print('type(123):', type(123))
print('type("str"):', type('str'))
print('type(None):', type(None))

# 比较类型
print('type(123) == int:', type(123) == int)
print('type("str") == str:', type('str') == str)
```

## 使用 isinstance()

```python
# isinstance 可以判断继承关系
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

print('isinstance(dog, Dog):', isinstance(dog, Dog))  # True
print('isinstance(dog, Animal):', isinstance(dog, Animal))  # True
print('isinstance(123, int):', isinstance(123, int))  # True
print('isinstance("str", str):', isinstance('str', str))  # True

# 可以判断多种类型
print('isinstance(123, (int, str, float)):', isinstance(123, (int, str, float)))  # True
```

## 使用 dir()

```python
# 获取对象的所有属性和方法
print('dir("ABC"):', dir('ABC'))

# 获取字符串长度
print('len("ABC"):', len('ABC'))
# 等价于
print('"ABC".__len__():', 'ABC'.__len__())

# 自定义类
class MyObject:
    def __init__(self):
        self.x = 9
    
    def power(self):
        return self.x * self.x

obj = MyObject()
print('dir(obj):', dir(obj))
```

## 使用 getattr、setattr、hasattr

```python
class MyObject:
    def __init__(self):
        self.x = 9

obj = MyObject()

# 检查是否有属性
print('hasattr(obj, "x"):', hasattr(obj, 'x'))  # True
print('hasattr(obj, "y"):', hasattr(obj, 'y'))  # False

# 获取属性
print('getattr(obj, "x"):', getattr(obj, 'x'))  # 9
print('getattr(obj, "y", 404):', getattr(obj, 'y', 404))  # 404 (默认值)

# 设置属性
setattr(obj, 'y', 10)
print('obj.y:', obj.y)  # 10
print('hasattr(obj, "y"):', hasattr(obj, 'y'))  # True
```
