# 多线程

Python 使用 `threading` 模块创建多线程。

## 使用 Thread 类创建线程

```python
import threading
import time

# 线程执行的函数
def loop():
    print('thread %s is running...' % threading.current_thread().name)
    n = 0
    while n < 5:
        n = n + 1
        print('thread %s >>> %s' % (threading.current_thread().name, n))
        time.sleep(1)
    print('thread %s ended.' % threading.current_thread().name)

print('thread %s is running...' % threading.current_thread().name)

# 创建线程
t = threading.Thread(target=loop, name='LoopThread')
t.start()
t.join()

print('thread %s ended.' % threading.current_thread().name)
```

## 锁 Lock - 保证线程安全

```python
import threading
import time

# 假定这是你的银行存款
balance = 0
lock = threading.Lock()

def change_it(n):
    # 先存后取，结果应该为 0
    global balance
    balance = balance + n
    balance = balance - n

def run_thread(n):
    for i in range(100000):
        # 获取锁
        lock.acquire()
        try:
            change_it(n)
        finally:
            # 释放锁
            lock.release()

t1 = threading.Thread(target=run_thread, args=(5,))
t2 = threading.Thread(target=run_thread, args=(8,))
t1.start()
t2.start()
t1.join()
t2.join()

print('balance:', balance)  # 使用锁后，结果始终为 0
```

## 使用局部线程变量 ThreadLocal

```python
import threading

# 创建全局 ThreadLocal 对象
local_school = threading.local()

def process_student():
    std = local_school.student
    print('Hello, %s (in %s)' % (std, threading.current_thread().name))

def thread_process(name):
    print('thread %s is running...' % threading.current_thread().name)
    local_school.student = name  # 绑定到线程的局部变量
    process_student()
    print('thread %s ended.' % threading.current_thread().name)

t1 = threading.Thread(target=thread_process, args=('Alice',), name='Thread-A')
t2 = threading.Thread(target=thread_process, args=('Bob',), name='Thread-B')
t1.start()
t2.start()
t1.join()
t2.join()
```

## 小结

- 线程是操作系统直接支持的执行单元
- Python 使用 `threading` 模块创建线程
- 多线程共享内存，需要注意线程安全问题
- 使用 `Lock` 保证线程安全
- `ThreadLocal` 解决线程间数据独立的问题
