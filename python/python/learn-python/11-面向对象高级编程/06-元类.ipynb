# 使用元类

元类是创建类的类，可以动态地创建类。

```python
# type() 可以创建一个类
# type(类名，(父类...)，dict 方法绑定)

# 定义一个函数作为方法
def fn(self, name='World'):
    print('Hello, %s.' % name)

# 使用 type() 创建 Hello 类
Hello = type('Hello', (object,), dict(hello=fn))

# 创建实例
h = Hello()
h.hello()  # Hello, World.

print('type(Hello):', type(Hello))
print('type(h):', type(h))
```

## 元类的概念

```python
# 元类是创建类的类
# class -> instance (对象)
# metaclass -> class

# type 是最基本的元类
# type 是 object 的子类，object 的 type 也是 type
print('type(object):', type(object))
print('type(type):', type(type))
```

## 自定义元类

```python
# 自定义元类示例：将方法名转为小写
class LowercaseMeta(type):
    """将所有方法名转为小写的元类"""
    def __new__(cls, name, bases, attrs):
        new_attrs = {}
        for k, v in attrs.items():
            # 跳过特殊方法和元类自身方法
            if k.startswith('__') and k.endswith('__'):
                new_attrs[k] = v
            else:
                new_attrs[k.lower()] = v
        return type.__new__(cls, name, bases, new_attrs)

class MyClass(metaclass=LowercaseMeta):
    def HELLO(self):
        print('Hello')
    
    def SayHI(self):
        print('Hi')

obj = MyClass()
# 方法名被转为小写
obj.hello()  # Hello
obj.sayhi()  # Hi
# obj.HELLO()  # AttributeError
```
