# 定制类

演示如何通过重写特殊方法来自定义类的行为。

```python
class Chain(object):
    """
    链式调用类
    用于生成类似 REST API 的路径：/users/michael/repos
    """
    
    def __init__(self, path=''):
        self._path = path

    def __getattr__(self, path):
        """动态属性访问"""
        return Chain('%s/%s' % (self._path, path))

    def __call__(self, path):
        """使实例可调用"""
        return Chain('%s/%s' % (self._path, path))

    def __str__(self):
        """字符串表示"""
        return self._path

    __repr__ = __str__

# 测试
if __name__ == '__main__':
    print(Chain().users("Michal").repo)
```
