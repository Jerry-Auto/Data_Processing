# 使用__slots__

限制类实例可以绑定的属性，节省内存。

```python
# 默认情况下，类实例可以动态添加任何属性
class Student(object):
    pass

s = Student()
s.name = 'Michael'  # 动态添加属性
print('name:', s.name)

s.age = 25
print('age:', s.age)
```

```python
# 使用__slots__限制可以添加的属性
class Student(object):
    __slots__ = ('name', 'age', 'score')  # 只允许添加这些属性
    
    def __init__(self):
        pass

s = Student()
s.name = 'Michael'
s.age = 25
s.score = 90
print('name:', s.name)
print('age:', s.age)
print('score:', s.score)

# s.grade = 'A'  # AttributeError: 'Student' object has no attribute 'grade'
```

```python
# __slots__对子类不起作用
class GraduateStudent(Student):
    pass

g = GraduateStudent()
g.name = 'Bob'
g.thesis = 'AI Research'  # 可以添加任意属性
print('thesis:', g.thesis)
```

## 注意事项

- `__slots__`只对当前类有效，对子类无效
- 使用`__slots__` 可以节省内存（特别是创建大量实例时）
- 限制了动态添加属性的能力
