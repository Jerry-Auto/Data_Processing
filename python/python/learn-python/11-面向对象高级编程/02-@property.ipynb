# @property 装饰器

演示 `@property` 装饰器的用法，将方法转换为属性调用。

```python
class Screen(object):
    """屏幕类 - 使用@property管理属性"""

    @property
    def width(self):
        """获取宽度"""
        return self._width

    @width.setter
    def width(self, value):
        """设置宽度"""
        self._width = value

    @property
    def height(self):
        """获取高度"""
        return self._height

    @height.setter
    def height(self, value):
        """设置高度"""
        self._height = value
    
    @property
    def resolution(self):
        """只读属性：分辨率 = 宽×高"""
        return self._width * self.height

# 测试
s = Screen()
s.width = 1024
s.height = 768
print('resolution =', s.resolution)
if s.resolution == 786432:
    print('测试通过!')
else:
    print('测试失败!')
```
