# 枚举类

演示 Python 枚举类的用法，使用 `enum.Enum`。

```python
from enum import Enum, unique

@unique
class Gender(Enum):
    """性别枚举"""
    Male = 0
    Female = 1

class Student(object):
    """学生类"""
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

# 测试
if __name__ == '__main__':
    bart = Student('Bart', Gender.Male)
    if bart.gender == Gender.Male:
        print('测试通过!')
    else:
        print('测试失败!')
```
