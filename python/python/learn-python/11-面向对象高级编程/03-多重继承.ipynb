# 多重继承

一个类可以继承多个父类，从而获得所有父类的方法。

```python
# 定义多个父类
class Runnable(object):
    def run(self):
        print('Running...')

class Flyable(object):
    def fly(self):
        print('Flying...')

# 多重继承
class Bird(Runnable, Flyable):
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f'{self.name} is speaking')

# 测试
bird = Bird('Eagle')
bird.run()    # 继承自 Runnable
bird.fly()    # 继承自 Flyable
bird.speak()  # 自己的方法
```

## Mixin 模式

通过多重继承给类添加多个功能。

```python
# Mixin 示例
class Animal(object):
    pass

# 定义功能 Mixin
class Mammal(Animal):
    """哺乳类"""
    pass

class Bird(Animal):
    """鸟类"""
    pass

class RunnableMixin(object):
    def run(self):
        print('Running...')

class FlyableMixin(object):
    def fly(self):
        print('Flying...')

# 组合使用
class Dog(Mammal, RunnableMixin):
    pass

class Bat(Mammal, FlyableMixin):
    pass

class Parrot(Bird, FlyableMixin):
    pass

# 测试
dog = Dog()
dog.run()

bat = Bat()
bat.fly()

parrot = Parrot()
parrot.fly()
```
