# 协程与 asyncio

Python 的异步 IO 编程。

## 协程概念

```python
# 协程是用户态的轻量级线程
# 通过 yield 实现协作式调度

def consumer():
    print('Consumer started')
    while True:
        n = yield
        print(f'Consumer received: {n}')

def producer(c):
    c.send(None)  # 启动生成器
    for i in range(5):
        print(f'Producer sending: {i}')
        c.send(i)
    c.close()

# 使用示例
c = consumer()
producer(c)

print('\n协程优势：单线程并发，无需锁，效率高')
```

## asyncio 基础

```python
import asyncio

# 异步函数
async def hello():
    print('Hello')
    await asyncio.sleep(1)  # 异步等待
    print('World')

# 运行异步函数
# asyncio.run(hello())

print('async/await 是 Python 3.5+ 的异步语法')
print('asyncio.run() 用于运行异步函数')
```

## 并发执行多个协程

```python
import asyncio
import time

async def task(name, delay):
    print(f'{name} 开始')
    await asyncio.sleep(delay)
    print(f'{name} 结束，耗时{delay}秒')
    return f'{name} 结果'

async def main():
    # 并发执行
    results = await asyncio.gather(
        task('任务 A', 1),
        task('任务 B', 2),
        task('任务 C', 3),
    )
    print('所有任务完成:', results)

# asyncio.run(main())

print('asyncio.gather() 用于并发执行多个协程')
```
