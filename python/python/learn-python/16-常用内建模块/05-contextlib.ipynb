# 上下文管理器

使用 contextlib 模块创建上下文管理器。

```python
from contextlib import contextmanager

@contextmanager
def tag(name):
    print(f'<{name}>')
    yield
    print(f'</{name}>')

with tag('h1'):
    print('Hello')
    print('World')
```
