# itertools 模块

提供用于创建迭代器的函数。

```python
import itertools

# 无限迭代器
count_iter = itertools.count(1, 2)  # 1, 3, 5, 7, ...
for i, n in enumerate(count_iter):
    if i >= 5:
        break
    print(n)
```

```python
# 有限迭代器
chain = itertools.chain('ABC', 'XYZ')
print('chain:', list(chain))  # ['A', 'B', 'C', 'X', 'Y', 'Z']
```

```python
# 分组迭代器
groupby = itertools.groupby('AAABBBCCCAAA')
for key, group in groupby:
    print(f'{key}: {list(group)}')
```

```python
# 排列组合
print('permutations:', list(itertools.permutations('ABC', 2)))
print('combinations:', list(itertools.combinations('ABC', 2)))
```
