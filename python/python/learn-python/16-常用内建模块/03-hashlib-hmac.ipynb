# hashlib 模块

哈希算法模块，用于计算消息摘要。

```python
import hashlib

# 计算 MD5
md5 = hashlib.md5()
md5.update('hello world'.encode('utf-8'))
print('MD5:', md5.hexdigest())

# 计算 SHA256
sha256 = hashlib.sha256()
sha256.update('hello world'.encode('utf-8'))
print('SHA256:', sha256.hexdigest())
```

## 分块计算哈希

```python
# 对于大文件，可以分块计算哈希
md5 = hashlib.md5()
md5.update(b'hello ')
md5.update(b'world')  # 结果与一次性计算相同
print('分块 MD5:', md5.hexdigest())
```

## 使用 HMAC

```python
import hmac
import hashlib

# 使用 HMAC-SHA256
message = b'Hello, World!'
key = b'secret-key'

h = hmac.new(key, message, hashlib.sha256)
print('HMAC:', h.hexdigest())
```
