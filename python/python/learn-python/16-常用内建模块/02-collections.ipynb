# collections 模块

Python 内建集合模块，提供额外的数据结构。

## namedtuple - 命名的元组

```python
from collections import namedtuple

# 创建具名元组类
Point = namedtuple('Point', ['x', 'y'])

# 使用
p = Point(1, 2)
print('x:', p.x)
print('y:', p.y)

# 判断是否是元组
print('isinstance(p, tuple):', isinstance(p, tuple))
```

## deque - 双向队列

```python
from collections import deque

# 创建双端队列
q = deque(['a', 'b', 'c'])

# 从两端添加元素
q.append('d')
q.appendleft('start')

# 从两端弹出元素
print('pop:', q.pop())
print('popleft:', q.popleft())

print('最终队列:', q)
```

## defaultdict - 默认值的字典

```python
from collections import defaultdict

# 普通 dict 访问不存在的 key 会报错
# dd = {}
# print(dd['missing'])  # KeyError

# defaultdict 可以设置默认值
dd = defaultdict(lambda: 'N/A')
dd['key1'] = 'value1'

print('key1:', dd['key1'])  # value1
print('key2:', dd['key2'])  # N/A (默认值)
```

## Counter - 计数器

```python
from collections import Counter

# 统计字符出现次数
c = Counter('programming')
print('字符统计:', c)
print('出现最多的 3 个字符:', c.most_common(3))
```

## OrderedDict - 有序的字典

```python
from collections import OrderedDict

# 普通 dict 不保证顺序（Python 3.7+ 保持插入顺序）
d = dict([('a', 1), ('b', 2), ('c', 3)])
print('普通 dict:', d)

# OrderedDict 保证顺序
od = OrderedDict([('a', 1), ('b', 2), ('c', 3)])
print('OrderedDict:', od)

# 按插入顺序迭代
print('keys:', list(od.keys()))
```
