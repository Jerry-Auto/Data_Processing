# 单元测试

演示 Python 单元测试的写法，使用 `unittest` 模块。

```python
import unittest

class Student(object):
    """学生类 - 根据成绩评级"""
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def get_grade(self):
        if self.score > 100:
            raise ValueError
        elif self.score >= 80:
            return 'A'
        elif self.score >= 60:
            return 'B'
        elif self.score >= 0:
            return 'C'
        else:
            raise ValueError

class TestStudent(unittest.TestCase):

    def test_80_to_100(self):
        """测试 80-100 分段"""
        s1 = Student('Bart', 80)
        s2 = Student('Lisa', 100)
        self.assertEqual(s1.get_grade(), 'A')
        self.assertEqual(s2.get_grade(), 'A')

    def test_60_to_80(self):
        """测试 60-80 分段"""
        s1 = Student('Bart', 60)
        s2 = Student('Lisa', 79)
        self.assertEqual(s1.get_grade(), 'B')
        self.assertEqual(s2.get_grade(), 'B')

    def test_0_to_60(self):
        """测试 0-60 分段"""
        s1 = Student('Bart', 0)
        s2 = Student('Lisa', 59)
        self.assertEqual(s1.get_grade(), 'C')
        self.assertEqual(s2.get_grade(), 'C')

    def test_invalid(self):
        """测试无效分数"""
        s1 = Student('Lisa', 101)
        s2 = Student('Bart', -1)

        with self.assertRaises(ValueError):
            s1.get_grade()

        with self.assertRaises(ValueError):
            s2.get_grade()

if __name__ == '__main__':
    unittest.main()
```
