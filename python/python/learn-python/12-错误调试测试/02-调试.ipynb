# 调试

Python 中常用的调试方法。

## 使用 print() 调试

```python
# 最简单的方法：用 print() 打印变量值
def foo(s):
    n = int(s)
    print(f'n = {n}')  # 打印变量值
    return 10 / n

foo('5')
```

## 使用 assert 断言

```python
# assert 语句：如果条件为 False，抛出 AssertionError
def foo(s):
    n = int(s)
    assert n != 0, 'n is zero!'  # 如果 n 为 0，抛出断言错误
    return 10 / n

print(foo('5'))

# 如果传入 0，会抛出 AssertionError
# foo('0')  # AssertionError: n is zero!
```

## 使用 logging 模块

```python
# logging 比 print() 更强大，可以记录日志到文件
import logging

# 设置日志级别：DEBUG, INFO, WARNING, ERROR, CRITICAL
logging.basicConfig(level=logging.DEBUG)

def foo(s):
    n = int(s)
    logging.info(f'n = {n}')  # 记录日志
    return 10 / n

foo('5')
```

## 使用 pdb 调试器

```python
# pdb 是 Python 的调试器模块
# 使用 pdb.set_trace() 设置断点

import pdb

def foo(s):
    n = int(s)
    pdb.set_trace()  # 运行到这里会自动暂停
    return 10 / n

# 运行到 set_trace() 后，可以在命令行中输入：
# - n 查看变量值
# - l 查看源代码
# - s 单步执行
# - c 继续执行
# - q 退出调试

# 注意：实际使用时需要注释掉下面的调用，然后在交互式环境中测试
# foo('5')
```

## 使用 pdb 进行单步调试

```python
# 使用 python -m pdb script.py 启动调试
# 常用命令：
# l - 查看当前代码附近的源代码
# n - 执行下一行代码
# s - 进入函数内部
# c - 继续执行直到下一个断点
# q - 退出调试器
# p 表达式 - 打印表达式的值

def calc(s):
    n = int(s)
    result = n * 2
    return result

# 在命令行运行：python -m pdb this_file.py
print(calc('5'))
```

## 小结

| 方法 | 优点 | 缺点 |
|------|------|------|
| print() | 简单直接 | 需要手动删除，输出混乱 |
| assert | 可以检查条件 | 只适合简单检查 |
| logging | 可以记录到文件，可控制级别 | 需要配置 |
| pdb | 功能强大，可单步调试 | 学习曲线较陡 |
