# 错误处理

演示 Python 的异常处理机制。

```python
from functools import reduce

def str2num(s):
    """字符串转数字"""
    return float(s)

def calc(exp):
    """计算表达式"""
    ss = exp.split('+')
    ns = map(str2num, ss)
    return reduce(lambda acc, x: acc + x, ns)

def main():
    try:
        r = calc('100 + 200 + 345')
        print('100 + 200 + 345 =', r)
        r = calc('99 + 88 + 7.6')
        print('99 + 88 + 7.6 =', r)
    except ValueError as e:
        print('ValueError:', e)
    except Exception as e:
        print('Exception:', e)

main()
```
