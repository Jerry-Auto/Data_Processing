# 模式匹配 match/case

Python 3.10+ 的新特性，可以针对某个变量匹配多种情况。

## 基本用法

```python
# 使用 match 语句匹配多个值
score = 'B'

match score:
    case 'A':
        print('score is A.')
    case 'B':
        print('score is B.')
    case 'C':
        print('score is C.')
    case _:
        print('invalid score.')  # _ 匹配任意值（默认情况）
```

## 复杂匹配

```python
# 多值匹配和条件匹配
age = 15

match age:
    case x if x < 10:
        print(f'< 10 years old: {x}')
    case 10:
        print('10 years old.')
    case 11 | 12 | 13 | 14 | 15 | 16 | 17 | 18:
        print('11~18 years old.')
    case 19:
        print('19 years old.')
    case _:
        print('not sure.')
```
