# dict 字典

键值对（key-value）存储，查找速度快。

```python
# 创建 dict
scores = {'Michael': 95, 'Bob': 75, 'Tracy': 85}
print('scores:', scores)

# 通过 key 获取 value
print("Michael's score:", scores['Michael'])

# 通过 key 添加或修改 value
scores['Adam'] = 90
print('after adding Adam:', scores)

scores['Bob'] = 80
print('after updating Bob:', scores)
```

```python
# 判断 key 是否存在
print('Michael in scores:', 'Michael' in scores)
print('Jack in scores:', 'Jack' in scores)

# 使用 get() 方法（不存在时返回 None 或指定值）
print('scores.get("Michael"):', scores.get('Michael'))
print('scores.get("Jack"):', scores.get('Jack'))
print('scores.get("Jack", -1):', scores.get('Jack', -1))

# 删除 key
scores.pop('Adam')
print('after pop Adam:', scores)
```

```python
# dict 的 key 必须是不可变对象
# key 不能是 list，因为 list 是可变的
# key = [1, 2, 3]  # TypeError: unhashable type: 'list'

# 但 key 可以是 tuple（如果 tuple 内元素也都是不可变的）
d = {(1, 2): 'value'}
print('tuple as key:', d)
```
