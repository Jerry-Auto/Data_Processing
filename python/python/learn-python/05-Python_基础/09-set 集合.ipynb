# set 集合

无序、不重复的元素集合。

```python
# 创建 set（需要传入 list）
s = set([1, 2, 3, 2, 3, 4])
print('set from list:', s)  # 自动去重：{1, 2, 3, 4}

# 添加元素
s.add(5)
print('after add 5:', s)

# 删除元素
s.remove(3)
print('after remove 3:', s)
```

```python
# set 的交集、并集、差集
s1 = set([1, 2, 3, 4])
s2 = set([3, 4, 5, 6])

print('s1:', s1)
print('s2:', s2)

# 交集
print('s1 & s2:', s1 & s2)  # {3, 4}

# 并集
print('s1 | s2:', s1 | s2)  # {1, 2, 3, 4, 5, 6}

# 差集（在 s1 中但不在 s2 中）
print('s1 - s2:', s1 - s2)  # {1, 2}
```
