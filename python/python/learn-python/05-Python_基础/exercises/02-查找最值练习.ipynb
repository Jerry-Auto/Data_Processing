# 列表练习 - 查找最大值和最小值

对应教程：5.3 使用 list

实现一个函数，查找列表中的最大值和最小值，并以元组形式返回。

```python
def findMinAndMax(l):
    """查找列表中的最大值和最小值"""
    if len(l) == 0:
        return None, None
    
    min_val = l[0]
    max_val = l[0]
    
    for val in l:
        if val < min_val:
            min_val = val
        if val > max_val:
            max_val = val
    
    return min_val, max_val

# 测试
if findMinAndMax([]) != (None, None):
    print('测试失败!')
elif findMinAndMax([7]) != (7, 7):
    print('测试失败!')
elif findMinAndMax([7, 1]) != (1, 7):
    print('测试失败!')
elif findMinAndMax([7, 1, 3, 9, 5]) != (1, 9):
    print('测试失败!')
else:
    print('测试成功!')
```
