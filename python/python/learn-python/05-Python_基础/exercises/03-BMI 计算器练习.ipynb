# 条件判断练习 - BMI 计算器

对应教程：5.4 条件判断

根据身高和体重计算 BMI 指数，并判断健康状况。

```python
# 小明身高 1.75m，体重 80.5kg
height = 1.75
weight = 80.5

# BMI = 体重 / 身高²
bmi = weight / height**2
print("bmi: %.3f" % bmi)

# 根据 BMI 判断健康状况
if bmi < 18.5:
    print("过轻")
elif bmi < 25:
    print("正常")
elif bmi < 28:
    print("过重")
elif bmi < 32:
    print("肥胖")
else:
    print("严重肥胖")
```
