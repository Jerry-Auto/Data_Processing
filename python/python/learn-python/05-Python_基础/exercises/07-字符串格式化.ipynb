# 字符串格式化详解

Python 中三种字符串格式化方法。

## % 格式化（旧式）

```python
name = 'Alice'
age = 25

# 基本用法
print('Hello, %s! You are %d years old.' % (name, age))

# 常用格式说明符
pi = 3.14159
num = 42

print('%d - 整数' % num)
print('%.2f - 保留 2 位小数' % pi)
print('%e - 科学计数法' % pi)
print('%x - 十六进制' % num)

# 字典格式化
person = {'name': 'Bob', 'age': 30}
print('%(name)s is %(age)d' % person)
```

## str.format() 方法

```python
# 位置参数
print('Hello, {}! You are {}.'.format('Alice', 25))

# 索引参数
print('{0} and {1}, {0} again'.format('A', 'B'))

# 关键字参数
print('Hello, {name}!'.format(name='Bob'))

# 格式化数字
num = 1234.5678
print('{:.2f}'.format(num))      # 保留 2 位小数
print('{:.2e}'.format(num))      # 科学计数法
print('{:%}'.format(0.1234))     # 百分比
print('{:,}'.format(1234567))    # 千位分隔符

# 对齐和填充
text = 'Python'
print('左对齐：{:<20}'.format(text))
print('右对齐：{:>20}'.format(text))
print('居中：{:^20}'.format(text))
print('填充：{:*^20}'.format(text))
```

## f-string（推荐）

```python
name = 'Alice'
age = 25
pi = 3.14159

# 基本用法
print(f'Hello, {name}! You are {age}.')

# 表达式
print(f'{age} * 2 = {age * 2}')

# 格式化
print(f'{pi:.2f}')
print(f'{1000000:,}')

# 调用方法
print(f'{name.upper()}')

# 多行 f-string
message = f"""
Name: {name}
Age: {age}
Pi: {pi:.2f}
"""
print(message)
```

## 三种方法对比

```python
name = 'Bob'
value = 3.14159

# % 格式化
print('%%: %s, %.2f' % (name, value))

# .format()
print('.format: {}, {:.2f}'.format(name, value))

# f-string (Python 3.6+, 最推荐)
print(f'f-string: {name}, {value:.2f}')
```
