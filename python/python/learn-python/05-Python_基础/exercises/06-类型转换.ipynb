# 类型转换

Python 中的数据类型转换方法。

## 转换为整数

```python
# int() 转换为整数
print(int('123'))       # 123
print(int(3.14))        # 3 (截断小数)
print(int('10', 2))     # 2 (二进制转十进制)
print(int('FF', 16))    # 255 (十六进制转十进制)
print(int(True))        # 1
print(int(False))       # 0
```

## 转换为浮点数

```python
# float() 转换为浮点数
print(float('3.14'))    # 3.14
print(float(5))         # 5.0
print(float('1e3'))     # 1000.0 (科学计数法)
```

## 转换为字符串

```python
# str() 转换为字符串
print(str(123))         # '123'
print(str(3.14))        # '3.14'
print(str([1, 2, 3]))   # '[1, 2, 3]'
print(str(True))        # 'True'
```

## 转换为列表

```python
# list() 转换为列表
print(list('hello'))    # ['h', 'e', 'l', 'l', 'o']
print(list((1, 2, 3)))  # [1, 2, 3]
print(list({'a', 'b'})) # ['a', 'b'] (集合转列表)
print(list(range(5)))   # [0, 1, 2, 3, 4]
```

## 转换为元组和集合

```python
# tuple() 转换为元组
print(tuple([1, 2, 3]))  # (1, 2, 3)
print(tuple('abc'))      # ('a', 'b', 'c')

# set() 转换为集合
print(set([1, 2, 2, 3])) # {1, 2, 3} (自动去重)
print(set('hello'))      # {'h', 'e', 'l', 'o'}
```

## 隐式类型转换

```python
# Python 会自动进行某些类型转换
result = 5 + 2.5  # 整数 + 浮点数 = 浮点数
print(f'5 + 2.5 = {result}, type: {type(result)}')

# 布尔值和整数的转换
print(f'True + 1 = {True + 1}')  # 2
print(f'False * 5 = {False * 5}')  # 0
```
