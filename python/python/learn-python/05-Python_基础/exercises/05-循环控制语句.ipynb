# 循环控制语句

break、continue 和 else 子句的使用。

## break - 退出循环

```python
# break 用于立即退出循环
for i in range(10):
    if i == 5:
        print(f'遇到{i}，退出循环')
        break
    print(f'当前数字：{i}')
```

## continue - 跳过本次循环

```python
# continue 用于跳过本次循环，继续下一次
for i in range(10):
    if i % 2 == 0:
        continue  # 跳过偶数
    print(f'奇数：{i}')
```

## 循环的 else 子句

```python
# else 在循环正常结束时执行（不被 break 中断）

# 正常结束 - 执行 else
for i in range(3):
    print(f'循环：{i}')
else:
    print('循环正常结束')

# 被 break 中断 - 不执行 else
for i in range(5):
    if i == 2:
        break
    print(f'循环：{i}')
else:
    print('这行不会被执行')
```

## 实用模式：查找元素

```python
# 查找第一个满足条件的元素
numbers = [1, 3, 5, 8, 9, 10]

# 找第一个偶数
for num in numbers:
    if num % 2 == 0:
        print(f'第一个偶数：{num}')
        break
else:
    print('没有找到偶数')
```

## 嵌套循环中的 break

```python
# break 只影响最内层循环
for i in range(3):
    print(f'外层：{i}')
    for j in range(5):
        if j == 2:
            print(f'  内层在 j={j} 时 break')
            break
        print(f'  内层：{j}')
```
