# 字符串处理练习 - 去除首尾空格

对应教程：5.2 字符串和编码

实现一个函数来去除字符串首尾的空格，不使用内置的 `strip()` 方法。

```python
# 方法一：双指针法
def myTrim(s):
    i = 0
    j = len(s) - 1
    while i < len(s) and s[i] == ' ':
        i += 1
    while j >= 0 and s[j] == ' ':
        j -= 1
    
    if i > j:
        return ''
    
    return s[i:j+1]

# 方法二：切片法（更简洁）
def trim(s):
    while s != '' and s[0] == ' ':
        s = s[1:]
    while s != '' and s[-1] == ' ':
        s = s[:-1]
    return s

# 测试
if trim('hello  ') != 'hello':
    print('测试失败!')
elif trim('  hello') != 'hello':
    print('测试失败!')
elif trim('  hello  ') != 'hello':
    print('测试失败!')
elif trim('  hello  world  ') != 'hello  world':
    print('测试失败!')
elif trim('') != '':
    print('测试失败!')
elif trim('    ') != '':
    print('测试失败!')
else:
    print('测试成功!')
```
