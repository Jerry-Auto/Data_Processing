# 运算符详解

Python 中的各种运算符及其用法。

## 算术运算符

```python
a, b = 10, 3

# 基本算术运算
print(f'加法：{a} + {b} = {a + b}')
print(f'减法：{a} - {b} = {a - b}')
print(f'乘法：{a} * {b} = {a * b}')
print(f'除法：{a} / {b} = {a / b}')  # 浮点除法
print(f'整除：{a} // {b} = {a // b}')  # 向下取整
print(f'取模：{a} % {b} = {a % b}')  # 余数
print(f'幂运算：{a} ** {b} = {a ** b}')
```

## 比较运算符

```python
x, y = 5, 10

print(f'x == y: {x == y}')  # 等于
print(f'x != y: {x != y}')  # 不等于
print(f'x > y: {x > y}')    # 大于
print(f'x < y: {x < y}')    # 小于
print(f'x >= y: {x >= y}')  # 大于等于
print(f'x <= y: {x <= y}')  # 小于等于
```

## 逻辑运算符

```python
a, b = True, False

print(f'a and b: {a and b}')  # 与
print(f'a or b: {a or b}')    # 或
print(f'not a: {not a}')      # 非
```

## 成员运算符

```python
lst = [1, 2, 3, 4, 5]

print(f'3 in lst: {3 in lst}')        # 检查是否存在
print(f'10 not in lst: {10 not in lst}')  # 检查是否不存在
```

## 位运算符

```python
a, b = 10, 3  # 二进制：1010, 0011

print(f'a & b: {a & b}')   # 按位与 0010 = 2
print(f'a | b: {a | b}')   # 按位或 1011 = 11
print(f'a ^ b: {a ^ b}')   # 按位异或 1001 = 9
print(f'~a: {~a}')         # 按位取反
print(f'a << 1: {a << 1}') # 左移 10100 = 20
print(f'a >> 1: {a >> 1}') # 右移 101 = 5
```

## 运算符优先级

```python
# 优先级从高到低：
# 1. 括号 ()
# 2. 幂运算 **
# 3. 一元运算符 + - ~
# 4. 算术运算符 * / // %
# 5. 算术运算符 + -
# 6. 比较运算符 < > <= >= == !=
# 7. 逻辑运算符 not and or

print(f'2 + 3 * 4 = {2 + 3 * 4}')        # 14 (先乘后加)
print(f'(2 + 3) * 4 = {(2 + 3) * 4}')    # 20 (括号优先)
print(f'2 ** 3 ** 2 = {2 ** 3 ** 2}')    # 512 (右结合)
print(f'(2 ** 3) ** 2 = {(2 ** 3) ** 2}')  # 64
```
