# 字符串和编码

Python 中的字符串处理以及字符编码。

## 字符串基础

```python
# 字符串可以用单引号或双引号表示
s1 = 'hello'
s2 = "world"
print('s1:', s1)
print('s2:', s2)

# 如果字符串内部有引号，可以用转义字符
s3 = 'I\'m "OK"!'
print('s3:', s3)

# 使用三引号表示多行字符串
s4 = '''line 1
line 2
line 3'''
print('s4:', s4)
```

## 转义字符

```python
# 常见转义字符
print('换行：hello\nworld')
print('制表符：hello\tworld')
print('反斜杠：hello\\world')

# 使用原始字符串（不转义）
print(r'原始字符串：hello\nworld')
```

## 字符串拼接和格式化

```python
# 字符串拼接
s1 = 'Hello'
s2 = 'World'
print('拼接:', s1 + ' ' + s2)

# 字符串重复
print('重复:', 'Hi' * 3)

# 格式化字符串
name = 'Michael'
age = 25

# 使用 f-string（推荐）
print(f'f-string: {name} is {age} years old')

# 使用 format()
print('format: {} is {} years old'.format(name, age))

# 使用 % 格式化
print('百分号：%s is %d years old' % (name, age))
```

## 字符串常用方法

```python
s = '  Hello World  '

# 去除首尾空格
print('strip():', s.strip())

# 转换为大写/小写
print('upper():', s.strip().upper())
print('lower():', s.strip().lower())

# 查找子串
print('find("World"):', s.find('World'))
print('index("World"):', s.index('World'))

# 替换
print('replace("World", "Python"):', s.replace('World', 'Python'))

# 分割
print('split():', s.strip().split(' '))

# 连接
print('join():', '-'.join(['A', 'B', 'C']))
```

## 字符编码

```python
# 字符转 ASCII 码
print('ord("A"):', ord('A'))  # 65
print('ord("中"):', ord('中'))  # 20013

# ASCII 码转字符
print('chr(65):', chr(65))  # A
print('chr(20013):', chr(20013))  # 中

# 字符串转 bytes
s = 'ABC'
print('encode:', s.encode('ascii'))  # b'ABC'

s = '中文'
print('encode:', s.encode('utf-8'))  # b'\xe4\xb8\xad\xe6\x96\x87'

# bytes 转字符串
b = b'\xe4\xb8\xad\xe6\x96\x87'
print('decode:', b.decode('utf-8'))  # 中文
```
