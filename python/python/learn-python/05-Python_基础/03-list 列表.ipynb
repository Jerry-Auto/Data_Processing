# list 列表

Python 内置的有序集合，可以随时添加和删除元素。

```python
# 创建 list
classmates = ['Michael', 'Bob', 'Tracy']
print('classmates:', classmates)

# 获取 list 长度
print('length:', len(classmates))

# 用索引访问元素（从 0 开始）
print('classmates[0]:', classmates[0])
print('classmates[1]:', classmates[1])
print('classmates[2]:', classmates[2])

# 使用负索引访问倒数元素
print('classmates[-1]:', classmates[-1])  # Tracy
print('classmates[-2]:', classmates[-2])  # Bob
```

```python
# 添加元素到末尾
classmates.append('Adam')
print('after append:', classmates)

# 插入元素到指定位置
classmates.insert(1, 'Jack')
print('after insert:', classmates)

# 删除末尾元素
classmates.pop()
print('after pop:', classmates)

# 删除指定位置元素
classmates.pop(1)
print('after pop(1):', classmates)

# 替换元素
classmates[1] = 'Sarah'
print('after replace:', classmates)
```

```python
# list 元素可以是不同类型
L = ['Apple', 123, True]
print('mixed list:', L)

# list 元素可以是另一个 list（二维数组）
s = ['python', 'java', ['asp', 'php'], 'scheme']
print('nested list:', s)
print('s[2][1]:', s[2][1])  # php

# 空 list
empty = []
print('empty list length:', len(empty))
```
