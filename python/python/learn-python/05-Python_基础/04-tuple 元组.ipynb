# tuple 元组

另一种有序列表，但一旦初始化就不能修改。

```python
# 创建 tuple
classmates = ('Michael', 'Bob', 'Tracy')
print('classmates:', classmates)

# 访问元素（和 list 一样）
print('classmates[0]:', classmates[0])
print('classmates[-1]:', classmates[-1])

# tuple 没有 append() 和 insert() 方法
# classmates.append('Adam')  # AttributeError
```

```python
# 定义空的 tuple
t = ()
print('empty tuple:', t)

# 定义只有 1 个元素的 tuple（必须加逗号消除歧义）
t = (1,)  # 注意逗号
print('single element tuple:', t)

# 如果不加逗号，就是数学括号
t = (1)
print('just a number:', t, type(t))
```

```python
# "可变的"tuple - 元素包含 list
t = ('a', 'b', ['A', 'B'])
print('before:', t)

# 修改 list 元素（tuple 本身没变，只是指向的 list 变了）
t[2][0] = 'X'
t[2][1] = 'Y'
print('after:', t)

# tuple 的"不变"是指每个元素的指向不变
```
