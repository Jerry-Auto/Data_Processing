# 列表生成式

演示列表生成式的用法，用简洁的方式生成列表。

```python
# 嵌套循环生成列表
l = [m + n for m in 'ABC' for n in 'XYZ']
print(l)
```

```python
# 条件过滤 + 字符串处理
l1 = ['Hello', 'World', 18, 'Apple', None]
l2 = [x.lower() for x in l1 if isinstance(x, str)]

print(l2)
if l2 == ['hello', 'world', 'apple']:
    print('测试通过!')
else:
    print('测试失败!')
```
