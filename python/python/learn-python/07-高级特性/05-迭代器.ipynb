# 迭代器

可以被 `next()` 函数调用并不断返回下一个值的对象。

```python
# 生成器都是迭代器
from collections.abc import Iterator

# 生成器
g = (x for x in range(5))
print('生成器是迭代器吗？', isinstance(g, Iterator))

# 可以使用 next() 迭代
print('next(g):', next(g))
print('next(g):', next(g))
print('next(g):', next(g))
```

```python
# list、dict、str 是可迭代对象，但不是迭代器
from collections.abc import Iterable

print('list 是 Iterable 吗？', isinstance([], Iterable))
print('list 是 Iterator 吗？', isinstance([], Iterator))

# 使用 iter() 将可迭代对象转换为迭代器
it = iter([1, 2, 3, 4, 5])
print('iter(list) 是 Iterator 吗？', isinstance(it, Iterator))

# 使用 next() 迭代
while True:
    try:
        print('next(it):', next(it))
    except StopIteration:
        break
```
