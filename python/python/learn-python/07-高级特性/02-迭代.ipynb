# 迭代

通过 for 循环遍历 dict、字符串等可迭代对象。

```python
# 迭代 dict 的 key
d = {'a': 1, 'b': 2, 'c': 3}
print('迭代 key:')
for key in d:
    print(key, ':', d[key])
```

```python
# 迭代 dict 的 value
print('\n迭代 value:')
for value in d.values():
    print(value)
```

```python
# 同时迭代 key 和 value
print('\n迭代 key 和 value:')
for k, v in d.items():
    print(k, ':', v)
```

```python
# 迭代字符串
print('\n迭代字符串:')
for ch in 'ABC':
    print(ch)
```

```python
# 使用 enumerate 获取索引和值
L = ['A', 'B', 'C']
print('\n使用 enumerate:')
for i, value in enumerate(L):
    print(i, ':', value)
```

```python
# 同时引用多个变量
L = [(1, 1), (2, 4), (3, 9)]
print('\n同时引用多个变量:')
for x, y in L:
    print(f'x={x}, y={y}')
```
