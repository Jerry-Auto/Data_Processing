## 标准维纳过程（布朗运动）

### 1. 定义

**标准维纳过程**（Standard Wiener Process），又称**标准布朗运动**，是连续时间随机过程 $\{W(t), t \ge 0\}$，满足以下四个条件：

1. **初始条件**：$W(0) = 0$（几乎必然）。

2. **独立增量**：对任意时间分割 $0 \le t_1 < t_2 < \cdots < t_n$，增量
   $$
   W(t_2)-W(t_1),\; W(t_3)-W(t_2),\;\dots,\;W(t_n)-W(t_{n-1})
   $$
   相互独立。

3. **平稳高斯增量**：对任意 $0 \le s < t$，增量服从正态分布：
   $$
   W(t) - W(s) \sim \mathcal{N}(0, t-s).
   $$
   即均值为 0，方差等于时间间隔长度。

4. **轨道连续**：以概率 1，函数 $t \mapsto W(t)$ 在 $[0,\infty)$ 上连续。

---

### 2. 基本性质

#### 2.1 矩性质
- **均值**：$\mathbb{E}[W(t)] = 0$
- **方差**：$\text{Var}(W(t)) = \mathbb{E}[W(t)^2] = t$
- **协方差**：$\mathbb{E}[W(t)W(s)] = \min(t, s)$

**推导协方差**：设 $s \le t$，利用独立增量性：
$$
\begin{aligned}
\mathbb{E}[W(t)W(s)] &= \mathbb{E}\big[(W(t)-W(s)+W(s))W(s)\big] \\
&= \mathbb{E}[(W(t)-W(s))W(s)] + \mathbb{E}[W(s)^2] \\
&= \mathbb{E}[W(t)-W(s)]\cdot\mathbb{E}[W(s)] + s \quad (\text{独立增量})\\
&= 0 \cdot 0 + s = s = \min(t,s).
\end{aligned}
$$

#### 2.2 自相似性（标度性）
对任意常数 $c > 0$，过程 $\{\frac{1}{\sqrt{c}}W(ct), t \ge 0\}$ 仍然是标准维纳过程。即：
$$
\frac{1}{\sqrt{c}}W(ct) \stackrel{d}{=} W(t).
$$
**含义**：在时间上拉伸，在空间上按平方根缩放，统计特性不变。

#### 2.3 时间反转对称性
- 过程 $\{tW(1/t), t > 0\}$（定义 $0 \cdot W(\infty) = 0$）也是标准维纳过程。
- 过程 $\{-W(t), t \ge 0\}$ 也是标准维纳过程。

#### 2.4 马尔可夫性
维纳过程是**连续时间马尔可夫过程**：给定当前值 $W(t)$，过去的值 $W(s), s < t$ 与未来的值 $W(u), u > t$ 相互独立。

#### 2.5 鞅性质
$W(t)$ 本身是**鞅**：对 $s < t$，有
$$
\mathbb{E}[W(t) \mid \mathcal{F}_s] = W(s),
$$
其中 $\mathcal{F}_s$ 表示到时刻 $s$ 为止的历史信息。

其他相关鞅：
- $W(t)^2 - t$ 是鞅
- $\exp(\theta W(t) - \frac{1}{2}\theta^2 t)$ 是指数鞅

---

### 3. 轨道性质

#### 3.1 连续但无处可微
以概率 1，维纳过程的样本轨道是处处连续但**处处不可微**的。这解释了为什么经典微积分不适用于处理维纳过程，而需要**伊藤微积分**。

#### 3.2 无界变差
维纳过程在任何有限区间上都具有**无界变差**。即轨道振荡极其剧烈，以至于“总路径长度”是无穷大。因此，积分 $\int f \, dW$ 不能按常规的 Riemann-Stieltjes 方式定义，需要伊藤积分。

#### 3.3 二次变差
设 $\Pi = \{t_0, t_1, \dots, t_n\}$ 是区间 $[0,T]$ 的一个划分，定义二次变差为：
$$
\lim_{\|\Pi\| \to 0} \sum_{i=1}^{n} \big( W(t_i) - W(t_{i-1}) \big)^2 = T \quad (\text{均方极限或几乎必然极限}).
$$
**与普通光滑函数的本质区别**：若 $f$ 是连续可微函数，其二次变差为零。维纳过程的二次变差非零（等于 $T$），这正是 $\sqrt{dt}$ 效应的来源——在伊藤引理中，$(dW)^2 = dt$ 是这一性质的微分形式。

#### 3.4 重对数律
以概率 1，
$$
\limsup_{t \to \infty} \frac{W(t)}{\sqrt{2t \log \log t}} = 1, \quad
\limsup_{t \to 0^+} \frac{W(t)}{\sqrt{2t \log \log(1/t)}} = 1.
$$
这个性质精确刻画了维纳过程轨道的振荡幅度。

### 4. 维纳过程与随机微分方程的关系

维纳过程是描述连续时间随机噪声的基本构件。物理、工程、金融中的随机扰动，通常以**增量** $dW(t)$ 的形式出现在随机微分方程（SDE）中：
$$
dX(t) = f(X(t), t)\,dt + \sigma(X(t), t)\,dW(t).
$$
这里，**$dW(t)$ 的方差尺度是 $\sqrt{dt}$**，这是理解伊藤积分和随机变分法的关键。

- **微分含义**：$dW(t)$ 不是普通微分，而是维纳过程在无穷小区间的增量，满足 $(dW(t))^2 = dt$（在伊藤意义上）。
- **物理意义**：代表大量微小独立随机冲击在极限下的总效应（中心极限定理 → 高斯分布 → $\sqrt{t}$ 尺度）。

---

### 5. 与其他概念的关系

| 概念 | 关系 |
|------|------|
| 离散时间随机游走 | 连续极限（时间步长 → 0，步长 → 0 按 $\sqrt{\Delta t}$ 尺度）即为维纳过程 |
| 布朗运动（物理） | 维纳过程是其严格的数学模型 |
| 高斯过程 | 维纳过程是特殊的高斯过程（零均值，核函数 $K(s,t) = \min(s,t)$） |
| 伊藤积分 | 维纳过程是积分器 $dW$ 的基础 |
| 随机最优控制 | 维纳过程进入状态方程作为噪声驱动项 |

---

### 6. 总结记忆卡片

- **定义四个条件**：起点为零 + 独立增量 + 高斯增量 $N(0, t-s)$ + 轨道连续
- **方差**：$\text{Var}(W(t)) = t$，协方差 $\min(s,t)$
- **核心尺度**：增量 $\sim \sqrt{dt}$，二次变差 $= T$
- **轨道**：连续但处处不可微，无界变差
- **角色**：连续时间随机过程的基本构件，SDE 的噪声源

维纳过程是你从确定性最优控制走向**随机最优控制**的第一块基石。在随机最优控制中，状态方程变为 SDE $dx = f(x,u)dt + \sigma(x,u)dW$，而这里所有的“随机性”都由 $dW$ 承载，伴随方程的推导也因伊藤公式中 $(dW)^2 = dt$ 的项而出现本质变化——这就是后续**随机最大值原理**和**HJB 方程**中二阶导数项的来源。
