# PMP

---

## 1. 问题设定

考虑受控系统
$$
\dot{x}(t) = f(x(t), u(t), t), \qquad x(t_0) = x_0,
$$
其中控制 $u(t)$ 取值于一个给定的非空集合 $U \subseteq \mathbb{R}^m$（通常为闭集，例如 $U = [0,1]$ 或有限离散点集）。代价泛函（求极小）为
$$
J(u) = \int_{t_0}^{t_f} L(x(t), u(t), t)\,dt + \Phi(x(t_f)).
$$
终端时间 $t_f$ 固定，终端状态自由。假设函数 $f, L, \Phi$ 对状态 $x$ 连续可微，对控制 $u$ 只需可测且有定义。

目标是：寻找最优控制 $u^*(\cdot): [t_0, t_f] \to U$，使 $J(u)$ 最小。

---

## 2. 经典变分法能做什么，不能做什么？

### 2.1 控制无约束时的经典方法

如果 $U = \mathbb{R}^m$（控制取值无任何限制），我们可以用拉格朗日乘子法处理动力学约束。引入**伴随变量**（也称协态变量）$\lambda(t) \in \mathbb{R}^n$，构造增广泛函
$$
\mathcal{L}(x,u,\lambda) = \int_{t_0}^{t_f} \left[ L(x,u,t) + \lambda^\top (f(x,u,t) - \dot{x}) \right] dt + \Phi(x(t_f)).
$$
将 $x, u, \lambda$ 都视为**独立函数**，分别对其做任意自由变分：

- 对 $\lambda$ 变分 $\to$ 恢复动力学约束 $\dot{x} = f(x,u,t)$
- 对 $x$ 变分（分部积分，利用 $\delta x(t_0)=0$）$\to$ 得到伴随方程
  $$
  \dot{\lambda} = -\frac{\partial H}{\partial x}, \qquad \lambda(t_f) = \left(\frac{\partial \Phi}{\partial x}(x(t_f))\right)^\top
  $$
  其中 $H(x,u,\lambda,t) := L(x,u,t) + \lambda^\top f(x,u,t)$ 称为哈密顿量。
- 对 $u$ 变分 $\to$ 得到控制方程
  $$
  \frac{\partial H}{\partial u} = 0.
  $$

这就是控制无约束时的经典必要条件：最优控制使哈密顿量达到驻点。

### 2.2 控制有界时，经典变分法为何失效？

当 $u(t) \in U$ 且 $U$ 是有界闭集（比如 $u \in [0,1]$）时，最优控制 $u^*(t)$ 很可能落在边界上（例如 $u^*(t) = 0$ 或 $u^*(t) = 1$）。

经典变分法要求 $\delta \mathcal{L} = 0$ 对**任意** $\delta u$ 成立，从而导出 $\frac{\partial H}{\partial u} = 0$。但这一推导有一个绝对的前提：

> **$\delta u$ 可以取任意方向、任意小的值。** 这要求最优控制 $u^*$ 必须位于允许集 $U$ 的**内部**。

一旦最优控制在边界上：
- 某些方向的扰动是非法的。例如 $u^* = 0$，$U = [0,1]$，则负方向的扰动 $-\delta u$ 会使控制变为负数，跑出允许集。
- 允许的变分不再构成整个空间，而是一个被边界截断的锥。
- 我们**不能**再要求一阶变分对所有 $\delta u$ 为零，只能要求对所有**允许方向**的变分，代价不下降。

因此，$\frac{\partial H}{\partial u} = 0$ 在边界点处通常不成立。例如：哈密顿量 $H = u$，控制集 $U = [0,1]$，最小值在 $u = 0$，但 $\frac{\partial H}{\partial u} = 1 \neq 0$。经典条件完全失效。

**经典变分法在控制有界时的失效，本质上是因为“变分的自由度被边界剥夺了”——你无法再围绕最优解做双向的无穷小试探。**

## 3. 针状变分的基本思想与构造

### 3.1 核心思想

既然围绕 $u^*$ 做任意小扰动这条路走不通，庞特里亚金换了一条路：

> **不要去扰动原来的控制函数 $u^*$，而是直接构造一个全新的、处处合法的试探控制 $u_\varepsilon$，然后拿它和 $u^*$ 比较代价。如果 $u^*$ 真的是最优的，那么任何这样的合法试探都不能让代价下降。**

这种方法不需要在 $u^*$ 附近做双向小扰动，因此不受控制集边界的限制。

### 3.2 针状变分的构造

任取一个时刻 $\tau \in [t_0, t_f)$ 和一个允许的控制值 $v \in U$。取充分小的 $\varepsilon > 0$，定义试探控制：
$$
\boxed{u_\varepsilon(t) = \begin{cases}
v, & t \in [\tau, \tau+\varepsilon], \\
u^*(t), & \text{其他}.
\end{cases}}
$$

这个构造的含义是：
- 在很短的时间区间 $[\tau, \tau+\varepsilon]$ 内，我们强行把控制从 $u^*(t)$ 切换为某个常值 $v$。
- 其余时间，控制保持原来的 $u^*(t)$ 不变。

**合法性的保证**：因为 $v \in U$，且 $u^*(t) \in U$ 本来就成立，所以拼接起来的 $u_\varepsilon(t)$ 处处属于 $U$。无论 $U$ 是闭集、离散集还是其他任意集合，这个构造都天然合法。

**关键特征**：
- $v$ 与 $u^*(\tau)$ 可以相差任意大。例如 $u^* = 0$，$v = 1$ 完全合法。
- 扰动幅度由控制值的变化 $v - u^*(\tau)$ 体现，而扰动的“小”体现在**时间长度 $\varepsilon \to 0$**上。
- 这完全不是经典意义上的“小扰动”，而是一种“脉冲式”的试探。

记扰动后的状态轨迹为 $x_\varepsilon(t)$，满足
$$
\dot{x}_\varepsilon(t) = f(x_\varepsilon(t), u_\varepsilon(t), t), \qquad x_\varepsilon(t_0) = x_0.
$$

## 4. 状态变分 $\delta x(t)$ 的演化分析

定义**状态变分**（一阶灵敏度）为：
$$
\delta x(t) \triangleq \lim_{\varepsilon \to 0^+} \frac{x_\varepsilon(t) - x^*(t)}{\varepsilon}.
$$
这是右极限，因为 $\varepsilon > 0$。

### 4.1 在 $t < \tau$ 时：状态轨迹完全重合

原因是，对状态变分，但是要满足状态转移方程(对协态的变分)，所以就是确定的转移

对于 $t < \tau$，根据构造 $u_\varepsilon(t) = u^*(t)$ 精确相等。两个轨迹满足完全相同的微分方程
$$
\dot{x}^*(t) = f(x^*(t), u^*(t), t), \quad \dot{x}_\varepsilon(t) = f(x_\varepsilon(t), u^*(t), t),
$$
且具有相同的初始条件 $x^*(t_0) = x_\varepsilon(t_0) = x_0$。由常微分方程初值问题解的唯一性，
$$
x_\varepsilon(t) = x^*(t) \quad (\text{精确相等}), \qquad \forall t \in [t_0, \tau].
$$
因此，
$$
\boxed{\delta x(t) = 0, \qquad t < \tau.}
$$

### 4.2 在 $t = \tau^+$ 处的跳跃条件

现在考察扰动刚结束的时刻 $\tau+\varepsilon$。分别写出两个状态在该时刻的值：
$$
\begin{aligned}
x^*(\tau+\varepsilon) &= x^*(\tau) + \int_{\tau}^{\tau+\varepsilon} f(x^*(s), u^*(s), s)\,ds, \\
x_\varepsilon(\tau+\varepsilon) &= x_\varepsilon(\tau) + \int_{\tau}^{\tau+\varepsilon} f(x_\varepsilon(s), v, s)\,ds.
\end{aligned}
$$
因为在 $\tau$ 时刻扰动尚未影响状态，有 $x_\varepsilon(\tau) = x^*(\tau)$。两式相减：
$$
x_\varepsilon(\tau+\varepsilon) - x^*(\tau+\varepsilon) = \int_{\tau}^{\tau+\varepsilon} \Big[ f(x_\varepsilon(s), v, s) - f(x^*(s), u^*(s), s) \Big] ds. \tag{4.1}
$$

被积函数在 $s \to \tau^+$ 时的极限为
$$
\lim_{s\to\tau^+} \big[ f(x_\varepsilon(s), v, s) - f(x^*(s), u^*(s), s) \big] = f(x^*(\tau), v, \tau) - f(x^*(\tau), u^*(\tau), \tau).
$$
（这里利用了 $f$ 的连续性，以及 $x_\varepsilon(s) \to x^*(\tau), x^*(s) \to x^*(\tau)$ 当 $s \to \tau^+$。）

由积分中值定理，对于连续函数 $F(s)$，在长度为 $\varepsilon$ 的区间上的积分满足
$$
\int_{\tau}^{\tau+\varepsilon} F(s)\,ds = \varepsilon F(\tau^+) + o(\varepsilon).
$$
应用到 (4.1) 得：
$$
x_\varepsilon(\tau+\varepsilon) - x^*(\tau+\varepsilon) = \varepsilon \big[ f(x^*(\tau), v, \tau) - f(x^*(\tau), u^*(\tau), \tau) \big] + o(\varepsilon). \tag{4.2}
$$

两边除以 $\varepsilon$，并取极限 $\varepsilon \to 0^+$：
$$
\lim_{\varepsilon \to 0^+} \frac{x_\varepsilon(\tau+\varepsilon) - x^*(\tau+\varepsilon)}{\varepsilon} = f(x^*(\tau), v, \tau) - f(x^*(\tau), u^*(\tau), \tau).
$$
左式正是 $\delta x(\tau^+)$（时间 $\tau+\varepsilon \to \tau^+$）。由此得到**跳跃条件**：
$$
\boxed{\delta x(\tau^+) = f(x^*(\tau), v, \tau) - f(x^*(\tau), u^*(\tau), \tau)}. \tag{4.3}
$$

**物理含义**：控制在 $\tau$ 处从 $u^*$ 跳变到 $v$，导致状态变化率瞬时不同，虽时间极短，但积累出一个 $O(\varepsilon)$ 的状态偏移。

### 4.3 在 $t > \tau$ 时的线性变分方程

现在考虑 $t > \tau$。对于充分小的 $\varepsilon$（使得 $\tau+\varepsilon < t$），在时刻 $t$ 两个轨迹都受控于 $u^*(t)$：
$$
\dot{x}_\varepsilon(t) = f(x_\varepsilon(t), u^*(t), t), \qquad \dot{x}^*(t) = f(x^*(t), u^*(t), t).
$$
两式相减：
$$
\frac{d}{dt}\big( x_\varepsilon(t) - x^*(t) \big) = f(x_\varepsilon(t), u^*(t), t) - f(x^*(t), u^*(t), t). \tag{4.4}
$$

对右边使用一阶泰勒展开（$f$ 对 $x$ 连续可微）：
$$
f(x_\varepsilon, u^*, t) = f(x^*, u^*, t) + \frac{\partial f}{\partial x}(x^*, u^*, t) \cdot (x_\varepsilon - x^*) + r(x_\varepsilon, x^*),
$$
其中余项 $r(x_\varepsilon, x^*) = o(\|x_\varepsilon - x^*\|)$ 当 $x_\varepsilon \to x^*$ 时。

代入 (4.4)：
$$
\frac{d}{dt}(x_\varepsilon - x^*) = \frac{\partial f}{\partial x}(x^*, u^*, t) \cdot (x_\varepsilon - x^*) + r. \tag{4.5}
$$

两边除以 $\varepsilon$：
$$
\frac{d}{dt} \left( \frac{x_\varepsilon - x^*}{\varepsilon} \right) = \frac{\partial f}{\partial x}(x^*, u^*, t) \cdot \left( \frac{x_\varepsilon - x^*}{\varepsilon} \right) + \frac{r}{\varepsilon}. \tag{4.6}
$$

现在取极限 $\varepsilon \to 0^+$。在标准 Lipschitz 假设下，可证对 $t \in [\tau, t_f]$，存在常数 $C$ 使得
$$
\|x_\varepsilon(t) - x^*(t)\| \le C\varepsilon.
$$
因此余项满足
$$
\frac{r}{\varepsilon} = \frac{o(\|x_\varepsilon - x^*\|)}{\varepsilon} \to 0 \qquad (\varepsilon \to 0^+).
$$
同时，在一定正则性条件下，极限与导数可交换：
$$
\lim_{\varepsilon\to 0^+} \frac{d}{dt} \left( \frac{x_\varepsilon - x^*}{\varepsilon} \right) = \frac{d}{dt} \lim_{\varepsilon\to 0^+} \left( \frac{x_\varepsilon - x^*}{\varepsilon} \right) = \frac{d}{dt} \delta x(t).
$$
对 (4.6) 取极限 $\varepsilon \to 0^+$，得到 $\delta x$ 满足的**线性变分方程**：
$$
\boxed{\frac{d}{dt} \delta x(t) = \frac{\partial f}{\partial x}\big(x^*(t), u^*(t), t\big) \cdot \delta x(t), \qquad t > \tau.} \tag{4.7}
$$
初始条件为 $\delta x(\tau^+)$ 由 (4.3) 给出。

**重要说明**：这个 $\delta x$ 不是我们自由选择的变分，而是由控制扰动通过动力学**被动确定**的响应。它描述了控制针状扰动引起的状态偏差在后续动力学中的传播规律。

## 5. 代价变化的精确计算

代价变化 $\Delta J = J(u_\varepsilon) - J(u^*)$ 为：
$$
\Delta J = \underbrace{\int_{\tau}^{\tau+\varepsilon} \big[L(x_\varepsilon(t), v, t) - L(x^*(t), u^*(t), t)\big]dt}_{\text{I}}
+ \underbrace{\int_{\tau+\varepsilon}^{t_f} \big[L(x_\varepsilon(t), u^*(t), t) - L(x^*(t), u^*(t), t)\big]dt}_{\text{II}}
+ \underbrace{\Phi(x_\varepsilon(t_f)) - \Phi(x^*(t_f))}_{\text{III}}.
$$

### 5.1 第一段 I：扰动窗口内的直接代价差

积分区间长度为 $\varepsilon$，被积函数在 $[\tau, \tau+\varepsilon]$ 上连续，由积分中值定理：
$$
\text{I} = \varepsilon \big[L(x^*(\tau), v, \tau) - L(x^*(\tau), u^*(\tau), \tau)\big] + o(\varepsilon). \tag{5.1}
$$
记 $\Delta L_0 := L(x^*(\tau), v, \tau) - L(x^*(\tau), u^*(\tau), \tau)$.

### 5.2 第二段 II 和第三段 III：状态偏差的间接影响

利用 $x_\varepsilon(t) = x^*(t) + \varepsilon \delta x(t) + o(\varepsilon)$，对运行代价 $L$ 线性展开：
$$
L(x_\varepsilon, u^*, t) - L(x^*, u^*, t) = \frac{\partial L}{\partial x}(x^*, u^*, t) \cdot (\varepsilon \delta x(t)) + o(\varepsilon).
$$
无穷小在有限区间上积分依旧是无穷小，因此，
$$
\begin{aligned}
\text{II} &= \int_{\tau+\varepsilon}^{t_f} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt \\
&= \varepsilon \int_{\tau+\varepsilon}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\, dt + o(\varepsilon).
\end{aligned}
$$
将积分下限从 $\tau+\varepsilon$ 改为 $\tau$，引入的误差为 $O(\varepsilon^2)$，可吸收进 $o(\varepsilon)$：
$$
\text{II} = \varepsilon \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\, dt + o(\varepsilon). \tag{5.2}
$$

对终端代价 $\Phi$ 线性展开：
$$
\Phi(x_\varepsilon(t_f)) - \Phi(x^*(t_f)) = \frac{\partial \Phi}{\partial x}(x^*(t_f)) \cdot (\varepsilon \delta x(t_f)) + o(\varepsilon). \tag{5.3}
$$

### 5.3 总代价变化的一阶主部

将 (5.1), (5.2), (5.3) 相加，提取因子 $\varepsilon$，忽略 $o(\varepsilon)$：
$$
\boxed{\Delta J = \varepsilon \left[ \Delta L_0 + \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\, dt + \frac{\partial \Phi}{\partial x(t_f)} \delta x(t_f) \right] + o(\varepsilon).} \tag{5.4}
$$

**结构分析**：
- 第一项 $\Delta L_0$：控制改变在扰动窗口内产生的直接代价差。
- 第二、三项：控制扰动引起的状态偏差 $\delta x$ 沿着后续轨迹累积的间接代价影响。这是一个从 $\tau$ 到 $t_f$ 的积分，包含了所有未来时刻的效应。

## 6. 伴随变量的引入——消除状态变分的积分

在针状变分推导中，我们之所以“引入”伴随变量 $\lambda(t)$ 并要求它满足协态方程，**本质是在使用经典变分法中处理等式约束的拉格朗日乘子法**。

我们面对的问题是：
- 控制扰动引起了状态偏差 $\delta x$，这个偏差沿着动力学传播，在代价中累积出一个积分项。
- 我们想把整个未来轨迹上的累积影响，折叠成扰动时刻的一个瞬时表达。

**怎么折叠？** 用拉格朗日乘子把动力学约束引入代价，然后对状态做变分。

---

## 完整的逻辑链

我们考虑一个增广的代价泛函：把动力学约束 $\dot{x} = f(x,u,t)$ 用乘子 $\lambda(t)$ 引入，构造

$$
\mathcal{L} = \int_{t_0}^{t_f} \big[ L(x,u,t) + \lambda^\top (f(x,u,t) - \dot{x}) \big] dt + \Phi(x(t_f)).
$$

这个泛函中，$x$、$u$、$\lambda$ 都被视为**独立变量**。

---

### 第一步：对 $\lambda$ 做变分

对 $\lambda$ 做任意变分 $\delta\lambda$：

$$
\delta_{\lambda}\mathcal{L} = \int_{t_0}^{t_f} \delta\lambda^\top (f - \dot{x})\,dt.
$$

令其为零（对任意 $\delta\lambda$），得到：

$$
\boxed{\dot{x} = f(x,u,t)}.
$$

这就是**状态方程**。它恢复了动力学约束。

---

### 第二步：对 $x$ 做变分

对 $x$ 做任意变分 $\delta x$（满足 $\delta x(t_0)=0$，因为初值固定）：

$$
\delta_x\mathcal{L} = \int_{t_0}^{t_f} \left[ \frac{\partial L}{\partial x}\delta x + \lambda^\top \frac{\partial f}{\partial x}\delta x - \lambda^\top \dot{\delta x} \right] dt + \frac{\partial \Phi}{\partial x(t_f)}\delta x(t_f).
$$

对最后一项 $-\lambda^\top \dot{\delta x}$ 做分部积分：

$$
-\int_{t_0}^{t_f} \lambda^\top \dot{\delta x}\,dt = -\big[\lambda^\top \delta x\big]_{t_0}^{t_f} + \int_{t_0}^{t_f} \dot{\lambda}^\top \delta x\,dt.
$$

因为 $\delta x(t_0)=0$，边界项只剩 $-\lambda(t_f)^\top \delta x(t_f)$。

代回：

$$
\delta_x\mathcal{L} = \int_{t_0}^{t_f} \left[ \frac{\partial L}{\partial x} + \lambda^\top \frac{\partial f}{\partial x} + \dot{\lambda}^\top \right] \delta x\,dt + \left( \frac{\partial \Phi}{\partial x(t_f)} - \lambda(t_f)^\top \right) \delta x(t_f).
$$

令其为零对任意 $\delta x$ 成立，得到：

1. **协态方程**（在积分号内）：
$$
\dot{\lambda}^\top = -\frac{\partial L}{\partial x} - \lambda^\top \frac{\partial f}{\partial x}
\quad\Longrightarrow\quad
\boxed{\dot{\lambda} = -\left(\frac{\partial L}{\partial x}\right)^\top - \left(\frac{\partial f}{\partial x}\right)^\top \lambda.}
$$

2. **终端条件**（边界项）：
$$
\boxed{\lambda(t_f)^\top = \frac{\partial \Phi}{\partial x(t_f)}.}
$$

定义哈密顿量 $H = L + \lambda^\top f$，则协态方程可简写为：

$$
\boxed{\dot{\lambda} = -\frac{\partial H}{\partial x}.}
$$

---

### 第三步：回到针状变分的上下文

现在，我们手里已经有了：
- 状态方程 $\dot{x}^* = f(x^*, u^*, t)$，
- 伴随方程 $\dot{\lambda} = -\frac{\partial H}{\partial x}$，
- 终端条件 $\lambda(t_f) = \nabla_x\Phi(x^*(t_f))^\top$.

**这些不是“规定”，而是对增广泛函独立变分 $x$ 和 $\lambda$ 的自然结果。**

接下来，针状变分推导中需要用到 $\lambda^\top \delta x$ 的性质。我们刚才从变分原理导出的伴随方程，正好保证了：

$$
\frac{d}{dt}(\lambda^\top \delta x) = -\frac{\partial L}{\partial x}\delta x.
$$


我们面临的核心困难是：$\Delta J$ 中含有一个从 $\tau$ 到 $t_f$ 的积分 $\int \frac{\partial L}{\partial x} \delta x \, dt$，要计算它需要知道 $\delta x(t)$ 在整个未来区间上的演化。我们希望通过某种方式，把这一积分转化为只依赖于扰动时刻 $\tau$ 的瞬时量。

### 6.1 伴随变量的定义

引入**伴随变量**（协态）$\lambda: [t_0, t_f] \to \mathbb{R}^n$，它是一个待定的函数。定义**哈密顿量**：
$$
\boxed{H(x, u, \lambda, t) := L(x, u, t) + \lambda^\top f(x, u, t).}
$$

我们希望 $\lambda$ 能够“吸收” $\delta x$ 的积分。为此，计算 $\lambda^\top \delta x$ 对时间的全导数（$t > \tau$）：
$$
\frac{d}{dt}\big( \lambda(t)^\top \delta x(t) \big) = \dot{\lambda}(t)^\top \delta x(t) + \lambda(t)^\top \dot{\delta x}(t).
$$
代入 $\dot{\delta x} = \frac{\partial f}{\partial x} \delta x$（由 (4.7)）：
$$
\frac{d}{dt}(\lambda^\top \delta x) = \dot{\lambda}^\top \delta x + \lambda^\top \frac{\partial f}{\partial x} \delta x. \tag{6.1}
$$

计算哈密顿量对 $x$ 的偏导数：
$$
\frac{\partial H}{\partial x} = \frac{\partial L}{\partial x} + \lambda^\top \frac{\partial f}{\partial x}. \tag{6.2}
$$

如果我们**规定** $\lambda$ 满足
$$
\dot{\lambda}(t) = -\left(\frac{\partial H}{\partial x}\right)^\top = -\left(\frac{\partial L}{\partial x}\right)^\top - \left(\frac{\partial f}{\partial x}\right)^\top \lambda, \tag{6.3}
$$
那么代入 (6.1)：
$$
\begin{aligned}
\frac{d}{dt}(\lambda^\top \delta x) &= \left( -\frac{\partial L}{\partial x} - \lambda^\top \frac{\partial f}{\partial x} \right) \delta x + \lambda^\top \frac{\partial f}{\partial x} \delta x \\
&= -\frac{\partial L}{\partial x} \delta x.
\end{aligned}
$$
**完美！** 我们得到：
$$
\boxed{\frac{d}{dt}\big( \lambda^\top \delta x \big) = -\frac{\partial L}{\partial x} \delta x.} \tag{6.4}
$$

### 6.2 终端条件的确定

为了使 $\lambda$ 也能处理终端项，我们规定终端条件：
$$
\boxed{\lambda(t_f)^\top = \frac{\partial \Phi}{\partial x}(x^*(t_f)).} \tag{6.5}
$$
这样 $\lambda(t_f)^\top \delta x(t_f)$ 恰好等于 $\frac{\partial \Phi}{\partial x(t_f)} \delta x(t_f)$.

### 6.3 用伴随变量消去积分

对 (6.4) 从 $\tau^+$ 到 $t_f$ 积分：
$$
\lambda(t_f)^\top \delta x(t_f) - \lambda(\tau)^\top \delta x(\tau^+) = -\int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t) \, dt.
$$
利用终端条件 (6.5)，移项得：
$$
\boxed{\frac{\partial \Phi}{\partial x(t_f)} \delta x(t_f) + \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t) \, dt = \lambda(\tau)^\top \delta x(\tau^+).} \tag{6.6}
$$

**这就是伴随变量的核心作用**：它把状态偏差在整个未来轨迹上的累积效应（一个积分），精确地折叠成了扰动发生时刻的一个瞬时值 $\lambda(\tau)^\top \delta x(\tau^+)$。这个操作之所以可能，是因为 $\lambda(t)$ 的定义（倒向方程 + 终端条件）恰好编码了“未来代价对当前状态的敏感度”。

---

## 7. 导出逐点最小化条件

将 (6.6) 代入代价变化表达式 (5.4)：
$$
\begin{aligned}
\Delta J &= \varepsilon \Big[ \Delta L_0 + \lambda(\tau)^\top \delta x(\tau^+) \Big] + o(\varepsilon) \\
&= \varepsilon \Big[ L(v) - L(u^*) + \lambda(\tau)^\top \big( f(x^*(\tau), v, \tau) - f(x^*(\tau), u^*(\tau), \tau) \big) \Big] + o(\varepsilon).
\end{aligned}
$$
（这里 $L(v)$ 是 $L(x^*(\tau), v, \tau)$ 的简写，$L(u^*)$ 同理。）

利用哈密顿量定义 $H = L + \lambda^\top f$，两项完美合并：
$$
\boxed{\Delta J = \varepsilon \big[ H(x^*(\tau), v, \lambda(\tau), \tau) - H(x^*(\tau), u^*(\tau), \lambda(\tau), \tau) \big] + o(\varepsilon).} \tag{7.1}
$$

**解读**：针状扰动引起的代价变化的一阶主部，恰好是扰动值 $v$ 与最优值 $u^*(\tau)$ 在哈密顿量下的差值，乘以扰动时长 $\varepsilon$。

---

## 8. 庞特里亚金最小值原理

因为 $u^*$ 是最优控制，对任何合法的试探控制，代价都不能降低：$\Delta J \ge 0$。将 (7.1) 除以 $\varepsilon > 0$，并令 $\varepsilon \to 0^+$：
$$
H(x^*(\tau), v, \lambda(\tau), \tau) - H(x^*(\tau), u^*(\tau), \lambda(\tau), \tau) \ge 0, \qquad \forall v \in U.
$$
即：
$$
H(x^*(\tau), u^*(\tau), \lambda(\tau), \tau) \le H(x^*(\tau), v, \lambda(\tau), \tau), \qquad \forall v \in U.
$$

由于 $\tau \in [t_0, t_f)$ 和 $v \in U$ 都是**任意选取**的，这一不等式必须在几乎所有时刻都成立。因此：
$$
\boxed{u^*(t) = \arg\min_{v \in U} H(x^*(t), v, \lambda(t), t), \qquad \text{a.e. } t \in [t_0, t_f].}
$$

这就是**庞特里亚金最小值原理**（对于极小化问题。历史上也称最大值原理，只需对哈密顿量取负号即可互化）。

**关键特征**：
- 不依赖控制 $u$ 的可微性。
- 不要求控制集合 $U$ 是开集或凸集。
- 即使最优控制在边界上，条件依然成立（通过大小比较而非导数等零）。
- 若 $U$ 为开集且 $H$ 对 $u$ 可微，则退化为经典条件 $\frac{\partial H}{\partial u} = 0$.

---

## 9. 完整的必要条件总结

**庞特里亚金最小值原理**：若 $u^*(\cdot)$ 为最优控制，$x^*(\cdot)$ 为相应的最优状态轨迹，则存在伴随函数 $\lambda(\cdot)$ 满足：

1. **状态方程**（正向）：
   $$
   \dot{x}^*(t) = f(x^*(t), u^*(t), t), \qquad x^*(t_0) = x_0.
   $$

2. **伴随方程**（倒向）：
   $$
   \dot{\lambda}(t) = -\frac{\partial H}{\partial x}\big(x^*(t), u^*(t), \lambda(t), t\big), \qquad \lambda(t_f)^\top = \frac{\partial \Phi}{\partial x}(x^*(t_f)).
   $$
   其中 $H(x,u,\lambda,t) = L(x,u,t) + \lambda^\top f(x,u,t)$.

3. **最小值原理**（逐点）：
   $$
   H(x^*(t), u^*(t), \lambda(t), t) = \min_{v \in U} H(x^*(t), v, \lambda(t), t), \qquad \text{a.e. } t \in [t_0, t_f].
   $$

---

## 10. 方法对比与理论意义

| 方面 | 经典变分法 | 针状变分（庞特里亚金） |
|------|-----------|----------------------|
| 对控制的扰动 | 加一个函数扰动 $\delta u$，要求处处小 | 构造新试探控制 $u_\varepsilon$，仅在无穷小区间替换为 $v \in U$ |
| 对状态的处理 | $x$ 和 $u$ 同为独立变分变量 | $\delta x$ 是控制扰动的被动响应，由动力学唯一确定 |
| 合法性前提 | $u^*$ 必须在 $U$ 内部，允许双向扰动 | $v \in U$ 即可，$u^*$ 可在边界 |
| 控制条件 | $\frac{\partial H}{\partial u} = 0$（驻点） | $\min_{v \in U} H$（最小值） |
| 适用范围 | 控制无约束或仅等式约束 | 任意控制集合（闭集、离散集等） |

**针状变分的核心贡献**：
1. **绕开了边界限制**：用构造合法试探替代任意小扰动，使变分思想能够处理控制有界问题。
2. **统一了必要条件形式**：无论控制约束是什么类型，条件都统一为 $\min_{v \in U} H$。
3. **揭示了伴随变量的深层含义**：$\lambda(t)$ 是“未来代价对当前状态的梯度”，它将未来所有影响折叠回当前时刻。
4. **为数值方法奠基**：将无限维最优控制问题分解为每时刻的有限维优化，间接法和直接法均建立在这一原理之上。

这就是从经典变分法到庞特里亚金最小值原理的完整逻辑链条——用针状扰动突破控制约束壁垒，用伴随变量实现时间反演，最终将全局最优性条件转化为逐点哈密顿量最小化。

我们逐步拆解这个积分下限变换的完整数学细节。

---

## 1. 问题的起点

我们有：
$$
I_2 = \int_{\tau+\varepsilon}^{t_f} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt.
$$

目标是把积分下限从 $\tau+\varepsilon$ 换成 $\tau$，使得后续与伴随变量的结合更方便。需要严格证明这样做引入的误差是 $o(\varepsilon)$。

---

## 2. 将积分拆分为两部分

根据定积分的基本性质：
$$
\int_{\tau+\varepsilon}^{t_f} F(t)\,dt = \int_{\tau}^{t_f} F(t)\,dt - \int_{\tau}^{\tau+\varepsilon} F(t)\,dt.
$$

应用到我们的积分中，令 $F(t) = \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon)$：
$$
\begin{aligned}
I_2 &= \int_{\tau}^{t_f} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt - \int_{\tau}^{\tau+\varepsilon} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt \\
&= \varepsilon \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\,dt + o(\varepsilon) - \underbrace{\int_{\tau}^{\tau+\varepsilon} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt}_{\text{这一项需要证明是 } o(\varepsilon)}.
\end{aligned}
$$

---

## 3. 估计减去的那个小区间积分

关键是要估算：
$$
R := \int_{\tau}^{\tau+\varepsilon} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt.
$$

### 3.1 分析被积函数

- $\frac{\partial L}{\partial x}$ 沿最优轨迹是连续函数，因此在紧集 $[\tau, \tau+\varepsilon]$ 上有界。记 $M_1 = \max_{t \in [\tau, \tau+\varepsilon]} \|\frac{\partial L}{\partial x}\|$。
- $\delta x(t)$ 是线性微分方程 (4.7) 的解，在 $[\tau, \tau+\varepsilon]$ 上连续，因此也有界。记 $M_2 = \max_{t \in [\tau, \tau+\varepsilon]} \|\delta x(t)\|$。
- 因此，在 $[\tau, \tau+\varepsilon]$ 上，$\frac{\partial L}{\partial x} \delta x(t)$ 是有界的，设其上界为 $M = M_1 M_2$。

### 3.2 用绝对值的三角不等式估计

$$
\begin{aligned}
|R| &= \left| \int_{\tau}^{\tau+\varepsilon} \left[ \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right] dt \right| \\
&\le \int_{\tau}^{\tau+\varepsilon} \left| \varepsilon \frac{\partial L}{\partial x} \delta x(t) + o(\varepsilon) \right| dt \quad (\text{积分的绝对值 } \le \text{ 绝对值的积分}) \\
&\le \int_{\tau}^{\tau+\varepsilon} \left( \left| \varepsilon \frac{\partial L}{\partial x} \delta x(t) \right| + |o(\varepsilon)| \right) dt \quad (\text{三角不等式}) \\
&= \int_{\tau}^{\tau+\varepsilon} \left( \varepsilon \left| \frac{\partial L}{\partial x} \delta x(t) \right| \right) dt + \int_{\tau}^{\tau+\varepsilon} |o(\varepsilon)| dt.
\end{aligned}
$$

### 3.3 估计每一项

**第一项**：
$$
\int_{\tau}^{\tau+\varepsilon} \varepsilon \left| \frac{\partial L}{\partial x} \delta x(t) \right| dt \le \varepsilon M \int_{\tau}^{\tau+\varepsilon} dt = \varepsilon M \cdot \varepsilon = M \varepsilon^2.
$$
这是 $O(\varepsilon^2)$ 的量。

**第二项**：这里的 $o(\varepsilon)$ 是依赖于 $t$ 的函数，但由于它在整个区间上都是 $\varepsilon$ 的高阶无穷小，其绝对值积分可估计为：
$$
\int_{\tau}^{\tau+\varepsilon} |o(\varepsilon)| dt \le o(\varepsilon) \cdot \varepsilon = o(\varepsilon^2).
$$
（严格来说，存在函数 $h(t, \varepsilon)$ 满足 $\lim_{\varepsilon \to 0} \frac{h(t, \varepsilon)}{\varepsilon} = 0$，则 $\int h \, dt \le \varepsilon \cdot \sup |h| = o(\varepsilon^2)$。）

---

## 4. 最终结论

综合以上估计：
$$
R = O(\varepsilon^2) + o(\varepsilon^2) = o(\varepsilon).
$$

因为 $O(\varepsilon^2)$ 除以 $\varepsilon$ 后是 $O(\varepsilon) \to 0$，所以 $R = o(\varepsilon)$。

因此：
$$
I_2 = \varepsilon \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\,dt + o(\varepsilon) - \underbrace{R}_{o(\varepsilon)} = \varepsilon \int_{\tau}^{t_f} \frac{\partial L}{\partial x} \delta x(t)\,dt + o(\varepsilon).
$$

---

## 5. 直观理解

- 我们在积分号外面已经有一个 $\varepsilon$ 因子。
- 多出来的那个小区间 $[\tau, \tau+\varepsilon]$ 长度本身就是 $\varepsilon$。
- 被积函数是有限的（$O(1)$ 量级）。
- 所以这个小区间积分是 $\varepsilon \times O(1) \times \varepsilon = O(\varepsilon^2)$。
- 再乘以外面的 $\varepsilon$ 因子，就变成了 $O(\varepsilon^2)$。
- 而我们保留的主部是 $O(\varepsilon)$。$O(\varepsilon^2)$ 相对于 $O(\varepsilon)$ 就是可以忽略的高阶无穷小。

**一句话总结**：移动积分下限多出来的那个积分区间，其长度是 $\varepsilon$ 量级，而被积函数也是有界的，所以这个积分是 $\varepsilon$ 量级。再乘以外面的 $\varepsilon$ 因子，就变成了 $\varepsilon^2$ 量级，在 $\varepsilon \to 0$ 时是比 $\varepsilon$ 更高阶的无穷小，可以被吸收进 $o(\varepsilon)$。


## 1. 两个关键符号的严格定义

### 1.1 大 O 符号：$O(\varepsilon^2)$

$R = O(\varepsilon^2)$ 的意思是：存在常数 $C > 0$ 和 $\varepsilon_0 > 0$，使得对于所有 $0 < \varepsilon < \varepsilon_0$，都有
$$
|R| \le C \varepsilon^2.
$$

换句话说，当 $\varepsilon$ 足够小时，$R$ 的大小被 $\varepsilon^2$ 的常数倍控制住。

### 1.2 小 o 符号：$o(\varepsilon)$

$R = o(\varepsilon)$ 的意思是：
$$
\lim_{\varepsilon \to 0^+} \frac{R}{\varepsilon} = 0.
$$
也就是说，$R$ 趋于零的速度**比 $\varepsilon$ 快**。

---

## 2. 从 $O(\varepsilon^2)$ 推导出 $o(\varepsilon)$

已知 $R = O(\varepsilon^2)$，即存在 $C$ 使得 $|R| \le C \varepsilon^2$。

现在考察 $\frac{R}{\varepsilon}$：
$$
\left| \frac{R}{\varepsilon} \right| = \frac{|R|}{\varepsilon} \le \frac{C \varepsilon^2}{\varepsilon} = C \varepsilon.
$$
因此，
$$
0 \le \left| \frac{R}{\varepsilon} \right| \le C \varepsilon.
$$

当 $\varepsilon \to 0^+$ 时，右边 $C\varepsilon \to 0$。由夹逼定理，
$$
\lim_{\varepsilon \to 0^+} \frac{R}{\varepsilon} = 0.
$$

这正是 $R = o(\varepsilon)$ 的定义。

---

## 3. 直观理解

- **$O(\varepsilon^2)$** 意味着 $R$ 的数量级是 $\varepsilon^2$（比如 $3\varepsilon^2$、$0.5\varepsilon^2$）。
- **除以 $\varepsilon$ 后**，数量级变成 $\varepsilon$（比如 $3\varepsilon$、$0.5\varepsilon$）。
- **当 $\varepsilon \to 0$ 时**，这个结果趋于 0，所以原量 $R$ 是比 $\varepsilon$ 更高阶的无穷小，即 $o(\varepsilon)$.

---

## 4. 一个具体数值例子

假设 $R = 5\varepsilon^2$。那么：
$$
\frac{R}{\varepsilon} = 5\varepsilon.
$$
当 $\varepsilon = 0.1$ 时，$\frac{R}{\varepsilon} = 0.5$；当 $\varepsilon = 0.01$ 时，$\frac{R}{\varepsilon} = 0.05$；当 $\varepsilon \to 0$ 时，这个比值趋于 0，所以 $R = o(\varepsilon)$。

---

## 5. 为什么用 $o(\varepsilon)$ 而不是保留 $O(\varepsilon^2)$？

在代价变化的一阶展开中，我们有：
$$
\Delta J = \varepsilon \cdot [\text{主部}] + R.
$$
我们关心的是 $\frac{\Delta J}{\varepsilon}$ 的极限。因为
$$
\frac{\Delta J}{\varepsilon} = [\text{主部}] + \frac{R}{\varepsilon},
$$
而 $\frac{R}{\varepsilon} \to 0$，所以极限就是主部。无论 $R$ 是 $O(\varepsilon^2)$ 还是更小的量，只要它除以 $\varepsilon$ 后趋于 0，就不会影响极限值。因此我们把它统一记作 $o(\varepsilon)$，表示“不影响极限的垃圾项”。

**所以，$O(\varepsilon^2) \implies o(\varepsilon)$，这是大 O 和小 o 定义的直接推论。**


---

## 1. 两者定义的根本区别

### 小 o 的定义

$R = o(\varepsilon)$ 当 $\varepsilon \to 0^+$ 的严格含义是：

$$
\boxed{\lim_{\varepsilon \to 0^+} \frac{R(\varepsilon)}{\varepsilon} = 0.}
$$

这意味着 $R$ 趋于零的**速度比 $\varepsilon$ 快**。$R$ 本身必须是无穷小（即 $R \to 0$），因为如果 $R$ 不趋于零，比值不可能趋于零。

### 大 O 的定义

$R = O(\varepsilon^2)$ 当 $\varepsilon \to 0^+$ 的严格含义是：

$$
\boxed{\exists C>0, \; \exists \varepsilon_0>0, \; \text{使得 } 0<\varepsilon<\varepsilon_0 \text{ 时}, \; |R(\varepsilon)| \le C \varepsilon^2.}
$$

大 O **不要求 $R$ 是无穷小**。它只要求 $R$ 的绝对值被 $\varepsilon^2$ 的常数倍控制住。

**但在这里，我们的 $R$ 恰好也是无穷小**，因为它是 $\varepsilon$ 的函数且 $R \to 0$。所以在我们讨论的具体情境中，两者都是无穷小。但它们的**信息含量**不同。

---

## 2. 信息含量的区别：谁更精确？

- **$R = O(\varepsilon^2)$** 告诉我们：$R$ 的衰减速度**至少是 $\varepsilon^2$** 量级（可能更快）。
- **$R = o(\varepsilon)$** 告诉我们：$R$ 的衰减速度**比 $\varepsilon$ 快**（但具体是 $\varepsilon^{1.5}$ 还是 $\varepsilon^2$ 还是 $\varepsilon^3$，不知道）。

因此，**$O(\varepsilon^2)$ 是比 $o(\varepsilon)$ 更强的结论**。它给出了更精确的数量级信息。
